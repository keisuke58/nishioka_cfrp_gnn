```python
import os
import csv

# INPファイルを読み込む関数
def read_original_inp_file(filename):
    with open(filename, 'r') as file:
        lines = file.readlines()
    return lines

# 指定した層の要素番号を抽出する関数
def extract_elements_from_inp(lines, layer_number):
    # 対象層に対応するキーワードを動的に生成
    if layer_number in {5, 6, 10, 11, 15, 16}:  # MAT0の層
        keyword = f"*Elset, elset=LAYER{layer_number}_MAT0"
    elif layer_number in {2, 9, 12, 19}:  # MAT45の層
        keyword = f"*Elset, elset=LAYER{layer_number}_MAT45"
    elif layer_number in {3, 8, 13, 18}:  # MAT90の層
        keyword = f"*Elset, elset=LAYER{layer_number}_MAT90"
    elif layer_number in {4, 7, 14, 17}:  # MAT-45の層
        keyword = f"*Elset, elset=LAYER{layer_number}_MAT-45"
    else:
        raise ValueError(f"Unsupported layer number: {layer_number}")

    elements = []
    capture = False

    for line in lines:
        if keyword in line:
            capture = True
            continue
        if capture:
            # 次のセクションが始まったら停止
            if line.startswith("**") or line.startswith("*"):
                break
            # 要素番号をリストに追加
            elements.extend([int(num) for num in line.strip().replace(" ", "").split(",") if num.isdigit()])
    return elements

# 各層の開始頂点を設定
layer_start_vertices = {
    2: 1767, 3: 3503, 4: 5239, 5: 6975, 6: 8711, 7: 10447, 8: 12183,
    9: 13919, 10: 15655, 11: 19127, 12: 20863, 13: 22599, 14: 24335,
    15: 26071, 16: 27807, 17: 29543, 18: 31279, 19: 33015, 20: 34751
}

def generate_defect_vertices(start_vertex, num_rows, num_columns, defect_size, step_size):
    vertices = []
    for i in range(0, num_rows - defect_size + 1, step_size):  # 行方向の移動
        for j in range(0, num_columns - defect_size + 1, step_size):  # 列方向の移動
            vertex = start_vertex + i + j * num_rows  # 列優先の計算式
            vertices.append(vertex)
    return vertices

def generate_defects(vertices, num_rows, defect_size):
    defects = []
    for start in vertices:
        defect = []
        for j in range(defect_size):  # 列方向のループ
            for i in range(defect_size):  # 行方向のループ
                element = start + i + j * num_rows  # 列優先の計算式
                defect.append(element)
        defects.append(defect)
        # print(f"{len(defect)} 個の欠陥ブロックを生成しました。")
    return defects

def write_inp_file_with_defect_exclusion(original_lines, layer_number, defect_elements, output_filename):
    new_lines = []
    inside_target_layer = False
    inside_defect_section = False
    layer_processed = False  # レイヤーが既に処理されたかどうかのフラグを追加

    # 対応するキーワードを動的に生成
    if layer_number in {5, 6, 10, 11, 15, 16}:  # MAT0の層
        keyword = f"*Elset, elset=LAYER{layer_number}_MAT0"
    elif layer_number in {2, 9, 12, 19}:  # MAT45の層
        keyword = f"*Elset, elset=LAYER{layer_number}_MAT45"
    elif layer_number in {3, 8, 13, 18}:  # MAT90の層
        keyword = f"*Elset, elset=LAYER{layer_number}_MAT90"
    elif layer_number in {4, 7, 14, 17}:  # MAT-45の層
        keyword = f"*Elset, elset=LAYER{layer_number}_MAT-45"
    else:
        raise ValueError(f"Unsupported layer number: {layer_number}")

    i = 0
    while i < len(original_lines):
        line = original_lines[i]

        # *Elset, elset=Defect セクションを挿入
        if "*Elset, elset=Defect" in line:
            new_lines.append(line)
            # 欠陥要素を16個ごとに区切って出力
            for j in range(0, len(defect_elements), 16):
                defect_chunk = defect_elements[j:j + 16]
                defect_line = ", ".join(map(str, defect_chunk)) + "\n"
                new_lines.append(defect_line)
            inside_defect_section = True
            i += 1
            continue
        elif inside_defect_section and (line.startswith("**") or line.startswith("*")):
            inside_defect_section = False

        # 指定したレイヤーの要素セットを検出
        if keyword in line:
            if not layer_processed:
                new_lines.append(line)  # キーワード行を追加
                # 元のファイル内で、現在のセクションの終わりまでスキップ
                i += 1
                while i < len(original_lines):
                    line = original_lines[i]
                    if line.startswith("**") or line.startswith("*"):
                        break  # セクションの終わり
                    i += 1
                # 新しい要素リストを追加
                updated_elements = extract_elements_from_inp(original_lines, layer_number)
                remaining_elements = [el for el in updated_elements if el not in defect_elements]
                for j in range(0, len(remaining_elements), 16):  # 16個ごとに区切って出力
                    elements_chunk = remaining_elements[j:j + 16]
                    element_line = ", ".join(map(str, elements_chunk)) + "\n"
                    new_lines.append(element_line)
                layer_processed = True  # レイヤーを処理済みに設定
                continue  # 次の行に進む（すでにiは進んでいる）
            else:
                # 既に処理済みの場合は、セクションをスキップ
                i += 1
                while i < len(original_lines):
                    line = original_lines[i]
                    if line.startswith("**") or line.startswith("*"):
                        break
                    i += 1
                continue
        else:
            new_lines.append(line)
            i += 1

    # 新しいINPファイルとして書き出し
    with open(output_filename, 'w') as f:
        f.writelines(new_lines)

# 欠陥ブロックをCSVに保存する関数
def save_defects_to_csv(defects, filename, layer_number, folder):
    csv_path = os.path.join(folder, filename)
    with open(csv_path, mode='w', newline='') as file:
        writer = csv.writer(file)
        writer.writerow(["Layer Number", "Block Number", "Block Elements"])
        
        for i, defect in enumerate(defects, 1):
            writer.writerow([layer_number, i] + defect)

# run.shファイルを生成する関数
def generate_sh_file(layers_data, filename):
    with open(filename, 'w') as file:
        file.write("#!/bin/bash\n")
        for layer_data in layers_data:
            for defect_info in layer_data['defects']:
                file.write(f"abq2024 job={defect_info['filename']} cpus=1\n")

# メイン処理
def main():
    # 元の要素リストをINPファイルから取得
    original_inp_filename = "re_defect_original.inp"
    original_lines = read_original_inp_file(original_inp_filename)

    # 欠陥サイズを指定
    defect_size = 6 # 8x8の欠陥ブロックを生成
    num_rows = 28
    num_columns = 62
    step_size = defect_size  # 欠陥が重ならないように配置

    # 保存先フォルダを作成
    output_folder = f"Defect{defect_size}x{defect_size}_data_1"
    os.makedirs(output_folder, exist_ok=True)

    # フォルダ構造の作成
    inner_folder = os.path.join(output_folder, f"Defect{defect_size}x{defect_size}_inner2-10")
    outer_folder = os.path.join(output_folder, f"Defect{defect_size}x{defect_size}_outer11-19")
    os.makedirs(outer_folder, exist_ok=True)
    os.makedirs(inner_folder, exist_ok=True)

    layers_data_outer = []
    layers_data_inner = []

    # 各層に対して処理を実行
    for layer_number in range(2, 20):  # Layer 2からLayer 19までを処理
        print(f"\n--- Layer {layer_number} の処理を開始 ---")
        layer_start_vertex = layer_start_vertices[layer_number]
        layer_elements = extract_elements_from_inp(original_lines, layer_number)

        # フォルダの選択
        if layer_number <= 10:
            layer_folder = inner_folder
            layers_data = layers_data_inner
        else:
            layer_folder = outer_folder
            layers_data = layers_data_outer

        # 欠陥の開始頂点を生成
        vertices = generate_defect_vertices(
            layer_start_vertex,
            num_rows=num_rows,
            num_columns=num_columns,
            defect_size=defect_size,
            step_size=step_size
        )

        # 欠陥ブロックを生成
        defects = generate_defects(vertices, num_rows=num_rows, defect_size=defect_size)
        
        # 欠陥ブロック情報をCSVに保存
        save_defects_to_csv(defects, f"Layer{layer_number}_defects{defect_size}x{defect_size}_output.csv", layer_number, layer_folder)

        layer_data = {
            "layer": layer_number,
            "defects": []
        }

        for i, defect_block in enumerate(defects, 1):
            filename = f"defect_{defect_size}x{defect_size}_Layer{layer_number}_Block{i}"
            inp_filename = os.path.join(layer_folder, f"{filename}.inp")
            write_inp_file_with_defect_exclusion(original_lines, layer_number, defect_block, inp_filename)
            layer_data["defects"].append({"filename": filename})

        layers_data.append(layer_data)

    # シェルスクリプトファイルを生成
    outer_sh_filename = os.path.join(outer_folder, "run.sh")
    inner_sh_filename = os.path.join(inner_folder, "run.sh")
    generate_sh_file(layers_data_outer, outer_sh_filename)
    generate_sh_file(layers_data_inner, inner_sh_filename)

if __name__ == "__main__":
    main()
```
