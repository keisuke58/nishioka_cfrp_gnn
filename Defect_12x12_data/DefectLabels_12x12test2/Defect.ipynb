```python
import numpy as np

# データのロード
data = np.load('12x12DefectLabel1_L8B10.npy')

# 0でない要素の数をカウント
non_zero_count = np.count_nonzero(data)
print(f"0でない要素の数: {non_zero_count}")

# 0でない要素のインデックスを取得
non_zero_indices = np.nonzero(data)[0]
print("0でない要素のインデックス（位置）:")
print(non_zero_indices)

# インデックスをCSVに保存
np.savetxt('non_zero_indices.csv', non_zero_indices, delimiter=',', fmt='%d')
print("非ゼロ要素のインデックスを 'non_zero_indices.csv' に保存しました。")
```

```python
import numpy as np

# データのロード
data = np.load('12x12DefectLabel1_L8B10.npy')

# 0でない要素の数をカウント
non_zero_count = np.count_nonzero(data)
print(f"0でない要素の数: {non_zero_count}")

# 0でない要素のインデックスを取得（一次元配列の場合）
non_zero_indices = np.nonzero(data)[0]

# 0でない要素の値を取得
non_zero_values = data[non_zero_indices]

# インデックスと値を2列に結合
csv_data = np.column_stack((non_zero_indices, non_zero_values))

# CSVファイルに保存
np.savetxt('non_zero_elements.csv', csv_data, delimiter=',', header='Index,Value', comments='', fmt=['%d','%f'])
print("非ゼロ要素のインデックスと値を 'non_zero_elements.csv' に保存しました。")
```
