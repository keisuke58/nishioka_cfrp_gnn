```python
import numpy as np
import pandas as pd
import os

# ファイルリスト
npy_files = [
    'train_x_GATModel_20241008_133645.npy',
    'train_y_GATModel_20241008_133645.npy',
    'val_x_GATModel_20241008_133645.npy',
    'val_y_GATModel_20241008_133645.npy',
    'test_x_GATModel_20241008_133645.npy',
    'test_y_GATModel_20241008_133645.npy'
]

# ファイルパスのディレクトリ
data_dir = '/home/nishioka/GNN/Defect_12x12_data/trainindata_confirm1/'  # ディレクトリを指定してください

# 各.npyファイルをロードしてCSVに保存
for file_name in npy_files:
    # フルパスの生成
    file_path = os.path.join(data_dir, file_name)
    
    # .npyファイルの読み込み
    data = np.load(file_path)
    
    # データフレームに変換
    df = pd.DataFrame(data)
    
    # CSVファイル名の生成
    csv_file_name = file_name.replace('.npy', '.csv')
    csv_file_path = os.path.join(data_dir, csv_file_name)
    
    # CSVとして保存
    df.to_csv(csv_file_path, index=False)
    print(f'Saved {csv_file_name} successfully.')

```

```python
import numpy as np
import os

# ディレクトリパスの指定
data_dir = '/home/nishioka/GNN/Defect_12x12_data/trainindata_confirm1/'

# 確認したいファイルリスト
npy_files = [
    'train_pairsGATModel_20241008_142126.npy',
    'val_pairsGATModel_20241008_142126.npy',
    'test_pairsGATModel_20241008_142126.npy',
]

# 各ファイルのデータ形式を確認
data_shapes = {}
for file_name in npy_files:
    file_path = os.path.join(data_dir, file_name)
    try:
        # .npyファイルの読み込み
        data = np.load(file_path)
        # 形状とデータ型の確認
        data_shapes[file_name] = {'shape': data.shape, 'dtype': data.dtype}
    except Exception as e:
        data_shapes[file_name] = {'error': str(e)}

data_shapes
```

```python
import pandas as pd

# 各.npyファイルを読み込み、1つのCSVファイルに結合して保存
combined_data = []

for file_name in npy_files:
    file_path = os.path.join(data_dir, file_name)
    try:
        # .npyファイルの読み込み
        data = np.load(file_path)
        # ファイル名に応じてデータセットタイプを追加
        dataset_type = file_name.split('pairs')[0].rstrip('_')
        # 各行にデータセットタイプを追加
        data_with_type = np.column_stack((np.full(data.shape[0], dataset_type), data))
        combined_data.append(data_with_type)
    except Exception as e:
        print(f"Error loading {file_name}: {e}")

# データを結合
combined_data = np.vstack(combined_data)

# データフレームに変換
df = pd.DataFrame(combined_data, columns=['Dataset_Type', 'Data_File', 'Label_File'])

# CSVとして保存
csv_output_path = os.path.join(data_dir, 'combined_data_pairs142126_GATModel.csv')
df.to_csv(csv_output_path, index=False)

csv_output_path
```
