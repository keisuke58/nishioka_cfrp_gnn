```python
import os
import numpy as np
import matplotlib.pyplot as plt

# Defect_4x4_Normalized1 フォルダ内のファイルパスを取得
normalized_files = [f for f in os.listdir() if f.endswith('.npy')]

# ランダムにサンプルを選択して正しく正規化できているか確認
def check_normalization(file_path, ax):
    data = np.load(file_path)
    min_value = np.min(data)
    max_value = np.max(data)
    
    # ヒストグラムをプロット
    ax.hist(data, bins=30, alpha=0.7)
    ax.set_title(f"{file_path}\nMin: {min_value:.4f}, Max: {max_value:.4f}")
    ax.set_xlabel('値')
    ax.set_ylabel('頻度')

# ランダムに5ファイルをサンプリングして確認
np.random.seed(42)  # 再現性のためランダムシードを固定
random_files = np.random.choice(normalized_files, 5, replace=False)

# 5つのサブプロットを用意
fig, axes = plt.subplots(1, 5, figsize=(20, 4))

# 各ファイルの統計情報とヒストグラムを確認
for ax, file in zip(axes, random_files):
    check_normalization(file, ax)

plt.tight_layout()
plt.show()
```

```python
import os
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Defect_4x4_Normalized1 folder to get file paths
normalized_files = [f for f in os.listdir() if f.endswith('.npy')]

# Check normalization using KDE plot
def check_normalization_kde(file_path, ax):
    data = np.load(file_path)
    min_value = np.min(data)
    max_value = np.max(data)
    
    # KDE plot
    sns.kdeplot(data, ax=ax, fill=True, alpha=0.7)
    ax.set_title(f"{file_path}\nMin: {min_value:.4f}, Max: {max_value:.4f}")
    ax.set_xlabel('Value')
    ax.set_ylabel('Density')

# Randomly select 5 files to check
np.random.seed(42)
random_files = np.random.choice(normalized_files, 5, replace=False)

# Create 5 subplots
fig, axes = plt.subplots(1, 5, figsize=(20, 4))

# Check each file's stats and KDE plot
for ax, file in zip(axes, random_files):
    check_normalization_kde(file, ax)

plt.tight_layout()
plt.show()
```

```python
import seaborn as sns

# Check normalization using Box plot
def check_normalization_boxplot(file_path, ax):
    data = np.load(file_path)
    sns.boxplot(data, ax=ax)
    ax.set_title(f"{file_path}\nBoxplot")
    ax.set_xlabel('Value')

# Create 5 subplots
fig, axes = plt.subplots(1, 5, figsize=(20, 4))

# Check each file's Box plot
for ax, file in zip(axes, random_files):
    check_normalization_boxplot(file, ax)

plt.tight_layout()
plt.show()
```

```python
import os
import numpy as np
import matplotlib.pyplot as plt

# Defect_4x4_Normalized1 folder: Get the file paths of all .npy files
normalized_files = [f for f in os.listdir() if f.endswith('.npy')]

# Function to check normalization and plot histogram
def check_normalization(file_path, ax, color):
    data = np.load(file_path)
    min_value = np.min(data)
    max_value = np.max(data)
    
    # Plot histogram
    ax.hist(data, bins=30, alpha=0.7, color=color)
    
    # Add title and labels
    ax.set_title(f"{file_path}\nMin: {min_value:.4f}, Max: {max_value:.4f}", fontsize=10)
    ax.set_xlabel('Value', fontsize=12)
    ax.set_ylabel('Frequency', fontsize=12)
    
    # Add lines for min-max normalization limits (if relevant to your data)
    ax.axvline(0, color='green', linestyle='dashed', linewidth=2, label="Min (0)")
    ax.axvline(1, color='red', linestyle='dashed', linewidth=2, label="Max (1)")
    
    # Add grid and legend
    ax.grid(True)
    ax.legend(loc='upper right')

# Randomly sample 5 files for normalization check
np.random.seed(42)  # Fix the random seed for reproducibility
random_files = np.random.choice(normalized_files, 5, replace=False)

# Colors for each histogram
colors = ['blue', 'orange', 'purple', 'green', 'red']

# Create 5 subplots
fig, axes = plt.subplots(1, 5, figsize=(20, 4))

# Check and plot histograms for each file
for ax, file, color in zip(axes, random_files, colors):
    check_normalization(file, ax, color)

plt.tight_layout()
plt.show()
```
