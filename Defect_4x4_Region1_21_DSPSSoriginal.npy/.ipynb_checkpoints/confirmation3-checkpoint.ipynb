```python
import numpy as np

# データのロード
file_name = "nodefect_4x4_Elnode.npy"
data = np.load(file_name)

# データの標準化
mean = np.mean(data)
std = np.std(data)
standardized_data = (data - mean) / std

# 標準化後のデータを表示
print(standardized_data)

# 必要なら、標準化されたデータを保存
np.save("standardized_nodefect_4x4_Elnode.npy", standardized_data)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Load the numpy file
file_name = "nodefect_4x4_Elnode.npy"
data = np.load(file_name)

# Standardize the data
mean = np.mean(data)
std = np.std(data)
standardized_data = (data - mean) / std

# Plot the standardized data
plt.figure(figsize=(10, 6))
plt.plot(standardized_data, label="Standardized Data")
plt.title("Standardized Data Plot")
plt.xlabel("Index")
plt.ylabel("Standardized Value")
plt.legend()
plt.grid(True)

# Display the plot
plt.show()
```

```python
# Print mean and standard deviation of the standardized data
print("Mean of standardized data:", np.mean(standardized_data))
print("Standard deviation of standardized data:", np.std(standardized_data))
```
