```python
import numpy as np
import os
import csv

# npyファイルを確認してCSVにまとめる関数
def verify_and_save_to_csv(output_folder, num_layers, num_blocks, csv_filename):
    with open(csv_filename, mode='w', newline='') as csv_file:
        csv_writer = csv.writer(csv_file)
        csv_writer.writerow(["Filename", "Shape", "Unique Labels", "Non-zero Label Count", "Non-zero Label Indices"])

        for layer in range(2, num_layers + 1):
            for i in range(1, num_blocks + 1):
                filename = f"{output_folder}/Defect11Class_L{layer}B{i}.npy"
                if (layer - 2) * num_blocks + i <= 5:
                    print(f"Processing file: {filename}")  # デバッグ用出力
                if os.path.exists(filename):
                    try:
                        labels = np.load(filename)
                        if (layer - 2) * num_blocks + i <= 5:
                            print(f"Loaded labels shape: {labels.shape}")  # デバッグ用出力
                        
                        # Get unique labels and indices of non-zero rows
                        unique_labels = np.unique(labels, axis=0).tolist()  # Convert to list for easier CSV output
                        non_zero_indices = [index for index, row in enumerate(labels) if not np.all(row == 0)]
                        non_zero_count = len(non_zero_indices)
                        
                        if (layer - 2) * num_blocks + i <= 5:
                            print(f"Non-zero label count: {non_zero_count}")  # デバッグ用出力
                        
                        # Write to CSV
                        csv_writer.writerow([filename, labels.shape, unique_labels, non_zero_count, non_zero_indices])
                        
                        if (layer - 2) * num_blocks + i <= 5:
                            print(f"Data written for file: {filename}")  # デバッグ用出力
                    except Exception as e:
                        if (layer - 2) * num_blocks + i <= 5:
                            print(f"Error processing file {filename}: {e}")  # エラーデバッグ用出力
                        csv_writer.writerow([filename, "Error loading file", "-", "-", "-"])
                else:
                    if (layer - 2) * num_blocks + i <= 5:
                        print(f"File does not exist: {filename}")  # デバッグ用出力
                    csv_writer.writerow([filename, "File does not exist", "-", "-", "-"])

# 出力フォルダとCSVファイル名の設定
output_folder = "/home/nishioka/GNN/Defect11Class_OneHot_test1"
csv_filename = "11classverification_results.csv"
num_blocks = 90
num_layers = 19

# 確認を実行してCSVにまとめる
verify_and_save_to_csv(output_folder, num_layers, num_blocks, csv_filename)
```
