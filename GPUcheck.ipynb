```python
import torch

# モデルの定義 (簡単な例として GCNModel を使用)
class GCNModel(torch.nn.Module):
    def __init__(self):
        super(GCNModel, self).__init__()
        self.conv1 = torch.nn.Linear(4, 16)  # 入力: 4次元データ
        self.conv2 = torch.nn.Linear(16, 32)
        self.conv3 = torch.nn.Linear(32, 16)
        self.fc_defect = torch.nn.Linear(16, 1)  # 出力: 1次元

    def forward(self, x):
        x = torch.relu(self.conv1(x))
        x = torch.relu(self.conv2(x))
        x = torch.relu(self.conv3(x))
        return self.fc_defect(x)
        # return torch.relu(self.fc_defect(x))  # 最後の層にもReLUを適用

# GPUが利用可能か確認
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

# モデルを定義し、GPUに送る
model = GCNModel().to(device)

# ダミーデータを作成し、GPUに送る
data = torch.randn(3654, 4).to(device)

# 推論を実行
with torch.no_grad():  # 勾配計算は不要なので無効化
    output = model(data)

print("推論結果:", output)
```
