```python
import os
import numpy as np
import matplotlib.pyplot as plt

# Defect_4x4_Standardized1 フォルダ内のファイルパスを取得
standardized_files = [f for f in os.listdir() if f.endswith('.npy')]

# ランダムにサンプルを選択して正しく標準化できているか確認
def check_standardization(file_path, ax):
    data = np.load(file_path)
    mean = np.mean(data)
    std_dev = np.std(data)
    
    # ヒストグラムをプロット
    ax.hist(data, bins=30, alpha=0.7)
    ax.set_title(f"{file_path}\nMean: {mean:.4f}, Std: {std_dev:.4f}")
    ax.set_xlabel('値')
    ax.set_ylabel('頻度')

# ランダムに5ファイルをサンプリングして確認
np.random.seed(42)  # 再現性のためランダムシードを固定
random_files = np.random.choice(standardized_files, 5, replace=False)

# 5つのサブプロットを用意
fig, axes = plt.subplots(1, 5, figsize=(20, 4))

# 各ファイルの統計情報とヒストグラムを確認
for ax, file in zip(axes, random_files):
    check_standardization(file, ax)

plt.tight_layout()
plt.show()
```

```python
import os
import numpy as np
import matplotlib.pyplot as plt

# Defect_4x4_Standardized1 folder: Get the file paths of all .npy files
standardized_files = [f for f in os.listdir() if f.endswith('.npy')]

# Function to check standardization and plot histogram
def check_standardization(file_path, ax, color):
    data = np.load(file_path)
    mean_value = np.mean(data)
    std_value = np.std(data)
    
    # Plot histogram
    ax.hist(data, bins=30, alpha=0.7, color=color)
    
    # Add title and labels
    ax.set_title(f"{file_path}\nMean: {mean_value:.4f}, Std Dev: {std_value:.4f}", fontsize=10)
    ax.set_xlabel('Value', fontsize=12)
    ax.set_ylabel('Frequency', fontsize=12)
    
    # Add lines for standardization reference (mean=0, std=1)
    ax.axvline(0, color='green', linestyle='dashed', linewidth=2, label="Mean (0)")
    ax.axvline(1, color='red', linestyle='dashed', linewidth=2, label="Std Dev (+1)")
    ax.axvline(-1, color='red', linestyle='dashed', linewidth=2, label="Std Dev (-1)")
    
    # Add grid and legend
    ax.grid(True)
    ax.legend(loc='upper right')

# Randomly sample 5 files for standardization check
np.random.seed(42)  # Fix the random seed for reproducibility
random_files = np.random.choice(standardized_files, 5, replace=False)

# Colors for each histogram
colors = ['blue', 'orange', 'purple', 'green', 'red']

# Create 5 subplots
fig, axes = plt.subplots(1, 5, figsize=(20, 4))

# Check and plot histograms for each file
for ax, file, color in zip(axes, random_files, colors):
    check_standardization(file, ax, color)

plt.tight_layout()
plt.show()
```
