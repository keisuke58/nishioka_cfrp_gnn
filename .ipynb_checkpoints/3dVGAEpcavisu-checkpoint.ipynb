```python
import os
import time
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch_geometric.nn import GCNConv, VGAE
from torch_geometric.data import Data
from torch_geometric.loader import DataLoader
from sklearn.model_selection import train_test_split
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import datetime

if torch.cuda.is_available():
    print("GPU is available")
    print(f"Using device: {torch.cuda.get_device_name(0)}")
else:
    print("GPU is not available, using CPU")

device = torch.device("cuda:0" if torch.cuda.is_available() else "cpu")
timestamp = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")

# Random seed for reproducibility
torch.manual_seed(42)
np.random.seed(42)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(42)

# Data paths
standardized_data_folder = "/home/nishioka/GNN/Defect_4x4_Normalized1"
label_data_folder = "/home/nishioka/GNN/Defectlabel4x4/DefectLabels_4x4_test1"

# Load coordinate data
x_coords = np.load("/home/nishioka/GNN/BasicdataforGNN/x_2layer_normalized.npy")[:3654]
y_coords = np.load("/home/nishioka/GNN/BasicdataforGNN/y_2layer_normalized.npy")[:3654]
z_coords = np.load("/home/nishioka/GNN/BasicdataforGNN/z_2layer_normalized.npy")[:3654]

# Load edge information
edges = np.load("/home/nishioka/GNN/BasicdataforGNN/edges_2layer.npy")
edge_index = torch.tensor(edges.T, dtype=torch.long).to(device)

# Data and label files
data_files = [f for f in os.listdir(standardized_data_folder) if f.startswith("Normalized1_Defect4x4_ELNOD")]
label_files = [f for f in os.listdir(label_data_folder) if f.startswith("DefectLabel_L")]

# Function to extract layer and block information from filenames
def extract_layer_block(file_name):
    if file_name.startswith("0"):
        return (0, 0)  # No defect data
    try:
        layer_block_str = file_name.split("_")[-1].replace(".npy", "")
        layer_str = layer_block_str.split("L")[1].split("B")[0]
        block_str = layer_block_str.split("B")[1]
        layer = int(layer_str)
        block = int(block_str)
        return (layer, block)
    except (ValueError, IndexError):
        print(f"Invalid file name format: {file_name}")
        return None

# Prepare data pairs
data_label_pairs = {}
for data_file in data_files:
    layer_block = extract_layer_block(data_file)
    if layer_block:
        data_label_pairs[layer_block] = {"data": data_file}

for label_file in label_files:
    layer_block = extract_layer_block(label_file)
    if layer_block and layer_block in data_label_pairs:
        data_label_pairs[layer_block]["label"] = label_file

# Valid pairs and adding defect-free data
valid_pairs = [(v["data"], v["label"]) for k, v in data_label_pairs.items() if "label" in v]
defect_free_pair = ("/home/nishioka/GNN/BasicdataforGNN/Normalized1_nodefect_ElNOD.npy",
                    "/home/nishioka/GNN/BasicdataforGNN/DefectLabel_nodefect.npy")

for _ in range(10):
    valid_pairs.append(defect_free_pair)

# Prepare data
def prepare_data(pairs):
    sampled_data = []
    sampled_labels = []
    
    for data_file, label_file in pairs:
        data_file_path = os.path.join(standardized_data_folder, data_file)
        label_file_path = os.path.join(label_data_folder, label_file)

        # Load data and labels
        try:
            values = np.load(data_file_path)[:3654]
            label = np.load(label_file_path)[:3654]
        except Exception as e:
            print(f"Error loading data: {e}")
            continue
        
        # Create node features by combining coordinate and stress data
        node_features = np.vstack((x_coords, y_coords, z_coords, values)).T
        sampled_data.append(node_features)
        sampled_labels.append(label)
    
    if len(sampled_data) == 0 or len(sampled_labels) == 0:
        print("No valid data found in the pairs.")
        return None

    # Convert to tensor
    sampled_data = np.stack(sampled_data)
    sampled_labels = np.hstack(sampled_labels)

    x = torch.tensor(sampled_data.reshape(-1, 4), dtype=torch.float).to(device)
    y = torch.tensor(sampled_labels, dtype=torch.float).to(device)

    return x, y

all_x, all_y = prepare_data(valid_pairs)

# Create Data objects for each sample
all_data_list = []
num_all_samples = all_x.shape[0] // 3654
for i in range(num_all_samples):
    data_file = data_files[i % len(data_files)]  # 正しくファイル名を取得
    data_obj = Data(x=all_x[i * 3654:(i + 1) * 3654], edge_index=edge_index, y=all_y[i * 3654:(i + 1) * 3654])
    
    # setattrを使ってfile_name属性を追加
    setattr(data_obj, 'file_name', str(data_file))
    
    all_data_list.append(data_obj)

# Use DataLoader for memory efficiency
batch_size = 16
train_data, test_data = train_test_split(all_data_list, test_size=0.2, random_state=42)
train_loader = DataLoader(train_data, batch_size=batch_size, shuffle=True)
test_loader = DataLoader(test_data, batch_size=batch_size, shuffle=False)

# Define GVAE model class
class Encoder(torch.nn.Module):
    def __init__(self, in_channels, hidden_channels):
        super(Encoder, self).__init__()
        self.conv1 = GCNConv(in_channels, hidden_channels)
        self.bn1 = nn.BatchNorm1d(hidden_channels)  # Batch Normalization
        self.dropout = nn.Dropout(p=0.3)  # Dropout
        self.conv2 = GCNConv(hidden_channels, hidden_channels * 2)
        self.conv_mu = GCNConv(hidden_channels * 2, hidden_channels)
        self.conv_logstd = GCNConv(hidden_channels * 2, hidden_channels)

    def forward(self, x, edge_index):
        x = F.relu(self.bn1(self.conv1(x, edge_index)))
        x = self.dropout(x)
        x = F.relu(self.conv2(x, edge_index))
        return self.conv_mu(x, edge_index), self.conv_logstd(x, edge_index)

class Decoder(torch.nn.Module):
    def __init__(self, hidden_channels, out_channels):
        super(Decoder, self).__init__()
        self.linear = nn.Linear(hidden_channels, out_channels)

    def forward(self, z):
        return self.linear(z)

hidden_channels=256
learning_rate=0.001
weight_decay=0.00001

encoder = Encoder(in_channels=4, hidden_channels=hidden_channels).to(device)
decoder = Decoder(hidden_channels=hidden_channels, out_channels=4).to(device)
model = VGAE(encoder).to(device)
optimizer = torch.optim.Adam(list(encoder.parameters()) + list(decoder.parameters()), lr=learning_rate, weight_decay=weight_decay)
scheduler = torch.optim.lr_scheduler.StepLR(optimizer, step_size=50, gamma=0.5)  # Learning rate scheduler

def loss_fn(recon_x, x, mu, logstd):
    recon_loss = F.mse_loss(recon_x, x, reduction='mean')
    kl_divergence = -0.5 * torch.mean(1 + 2 * logstd - mu.pow(2) - (2 * logstd).exp())
    return recon_loss + kl_divergence

# Training loop (with 2 epochs)
epochs = 200
for epoch in range(epochs):
    model.train()
    total_loss = 0
    for data in train_loader:
        data = data.to(device)
        optimizer.zero_grad()
        z = model.encode(data.x, data.edge_index)
        recon_x = decoder(z)
        loss = F.mse_loss(recon_x, data.x)  # reconstruction loss
        loss.backward()
        optimizer.step()
        total_loss += loss.item()
    
    print(f"Epoch {epoch + 1}/{epochs}, Loss: {total_loss / len(train_loader)}")

# Clustering and visualization
layer_wise_encoded = {}
defect_free_encoded = []

with torch.no_grad():
    for data in test_loader:
        data = data.to(device)

        # データがリストかどうかを確認し、リストの場合は最初のファイル名を取得
        file_name = data.file_name[0] if isinstance(data.file_name, list) else data.file_name

        # VGAEのencodeメソッドを使って潜在変数zを取得
        z = model.encode(data.x, data.edge_index)

        # ファイル名から層番号を取得
        layer_block_info = extract_layer_block(file_name)
        if layer_block_info:
            layer_number, _ = layer_block_info

            if layer_number == 0:
                defect_free_encoded.append(z.cpu().numpy())
            else:
                if layer_number not in layer_wise_encoded:
                    layer_wise_encoded[layer_number] = []
                layer_wise_encoded[layer_number].append(z.cpu().numpy())

# 各層の潜在変数をnumpy配列に変換
for layer_number in layer_wise_encoded:
    layer_wise_encoded[layer_number] = np.vstack(layer_wise_encoded[layer_number])

if defect_free_encoded:
    defect_free_encoded = np.vstack(defect_free_encoded)
else:
    defect_free_encoded = np.array([])

# PCAで次元削減
if defect_free_encoded.size > 0:
    all_encoded = np.vstack(list(layer_wise_encoded.values()) + [defect_free_encoded])
else:
    all_encoded = np.vstack(list(layer_wise_encoded.values()))

pca = PCA(n_components=2)
pca.fit(all_encoded)

# 欠陥なしデータの可視化
def visualize_defect_free_data():
    defect_free_x, defect_free_y = prepare_data([defect_free_pair])

    if defect_free_x is not None and defect_free_y is not None:
        model.eval()
        with torch.no_grad():
            defect_free_encoded = model.encode(defect_free_x, edge_index)
            defect_free_encoded_np = defect_free_encoded.cpu().numpy()
        
        # PCA で次元削減
        defect_free_pca = pca.transform(defect_free_encoded_np)
        
        # 欠陥なしデータの可視化
        plt.figure(figsize=(10, 8))
        plt.scatter(defect_free_pca[:, 0], defect_free_pca[:, 1], color='black', label='Defect-Free Data', s=20, alpha=0.8)
        plt.title('Visualization of Defect-Free Data')
        plt.xlabel('PCA Component 1')
        plt.ylabel('PCA Component 2')
        plt.legend(loc='upper right')
        plt.show()
    else:
        print("No defect-free data found to visualize.")

# 欠陥なしデータの可視化呼び出し
visualize_defect_free_data()

# その他の層データのプロット
plt.figure(figsize=(10, 8))
colors = plt.get_cmap('tab20', len(layer_wise_encoded))

# 層ごとのデータをプロット
for idx, (layer_number, encoded_data) in enumerate(layer_wise_encoded.items()):
    z_pca = pca.transform(encoded_data)
    plt.scatter(z_pca[:, 0], z_pca[:, 1], label=f'Layer {layer_number}', color=colors(idx), s=5, alpha=0.3)

plt.title('Latent Space Visualization by Layer and Defect-Free Data')
plt.xlabel('PCA Component 1')
plt.ylabel('PCA Component 2')
plt.legend(loc='upper right')
plt.colorbar(label='Layer Number')
plt.show()

# モデルのパラメータ（state_dict）を保存する関数
def save_model_state(model, path):
    torch.save(model.state_dict(), path)
    print(f"Model state_dict saved to {path}")

# 保存先のパスを指定
model_save_path = f"/home/nishioka/GNN/GNNmodel/gvae_model_{timestamp}.pth"

# モデルのパラメータを保存
save_model_state(model, model_save_path)
```

```python
from mpl_toolkits.mplot3d import Axes3D

# PCAで3次元への次元削減
pca = PCA(n_components=3)
pca.fit(all_encoded)

# データ全体の範囲を計算する関数
def get_axis_limits(pca_data_list):
    all_data = np.vstack(pca_data_list)
    x_limits = (all_data[:, 0].min(), all_data[:, 0].max())
    y_limits = (all_data[:, 1].min(), all_data[:, 1].max())
    z_limits = (all_data[:, 2].min(), all_data[:, 2].max())
    return x_limits, y_limits, z_limits

# 欠陥なしデータと層ごとのデータをPCAに変換する関数
def safe_pca_transform(data):
    if len(data) > 0:  # データが空でないかを確認
        data = np.atleast_2d(data)  # 1次元データを2次元に変換
        return pca.transform(data)
    else:
        return np.array([])  # 空のデータの場合、処理をスキップ

# 欠陥なしデータと層ごとのデータをまとめてPCAした後に範囲を計算
all_pca_data = [safe_pca_transform(defect_free_encoded)] + [safe_pca_transform(encoded_data) for encoded_data in layer_wise_encoded.values()]
all_pca_data = [data for data in all_pca_data if len(data) > 0]  # 空のデータは削除
x_lim, y_lim, z_lim = get_axis_limits(all_pca_data)

# 欠陥なしデータの可視化
def visualize_defect_free_data_3d():
    defect_free_x, defect_free_y = prepare_data([defect_free_pair])

    if defect_free_x is not None and defect_free_y is not None:
        model.eval()
        with torch.no_grad():
            defect_free_encoded = model.encode(defect_free_x, edge_index)
            defect_free_encoded_np = defect_free_encoded.cpu().numpy()
        
        # 3次元PCAで次元削減
        defect_free_pca = safe_pca_transform(defect_free_encoded_np)
        
        if defect_free_pca.size > 0:  # 欠陥なしデータが空でない場合にプロット
            # 3Dプロットの準備
            fig = plt.figure(figsize=(10, 8))
            ax = fig.add_subplot(111, projection='3d')
            # 欠陥なしデータをプロット（透明度をさらに高くする）
            ax.scatter(defect_free_pca[:, 0], defect_free_pca[:, 1], defect_free_pca[:, 2], 
                       color='black', label='Defect-Free Data', s=5, alpha=0.05)  # alpha=0.05でさらに薄く表示
            ax.set_title('3D Visualization of Defect-Free Data')
            ax.set_xlabel('PCA Component 1')
            ax.set_ylabel('PCA Component 2')
            ax.set_zlabel('PCA Component 3')
            
            # PCAの軸の範囲を設定
            ax.set_xlim(x_lim)
            ax.set_ylim(y_lim)
            ax.set_zlim(z_lim)
            
            plt.legend(loc='upper right')
            plt.show()
        else:
            print("No defect-free data found to visualize.")
    else:
        print("No defect-free data found to visualize.")

# 欠陥なしデータの3次元可視化
visualize_defect_free_data_3d()

# その他の層データの3次元プロット
def visualize_layer_data_3d():
    fig = plt.figure(figsize=(10, 8))
    ax = fig.add_subplot(111, projection='3d')
    
    colors = plt.get_cmap('tab20', len(layer_wise_encoded))
    
    # 層番号でソートした順にプロット
    sorted_layers = sorted(layer_wise_encoded.items(), key=lambda x: x[0])
    
    # 各層ごとのデータを3次元プロット（透明度を高くして、サイズをさらに小さく）
    for idx, (layer_number, encoded_data) in enumerate(sorted_layers):
        z_pca = safe_pca_transform(encoded_data)
        if z_pca.size > 0:  # データが空でない場合のみプロット
            ax.scatter(z_pca[:, 0], z_pca[:, 1], z_pca[:, 2], 
                       label=f'Layer {layer_number}', color=colors(idx), s=5, alpha=0.05)  # alpha=0.05でさらに薄く表示
    
    ax.set_title('3D Latent Space Visualization by Layer and Defect-Free Data')
    ax.set_xlabel('PCA Component 1')
    ax.set_ylabel('PCA Component 2')
    ax.set_zlabel('PCA Component 3')
    
    # PCAの軸の範囲を設定
    ax.set_xlim(x_lim)
    ax.set_ylim(y_lim)
    ax.set_zlim(z_lim)
    
    # 凡例の順番を層番号で統一
    plt.legend(loc='upper right')
    plt.show()

# 層ごとの3次元可視化
visualize_layer_data_3d()

```

```python
from mpl_toolkits.mplot3d import Axes3D
from sklearn.cluster import KMeans
from sklearn.cluster import DBSCAN

# PCAで3次元への次元削減
pca = PCA(n_components=3)
pca.fit(all_encoded)

# データ全体の範囲を計算する関数
def get_axis_limits(pca_data_list):
    all_data = np.vstack(pca_data_list)
    x_limits = (all_data[:, 0].min(), all_data[:, 0].max())
    y_limits = (all_data[:, 1].min(), all_data[:, 1].max())
    z_limits = (all_data[:, 2].min(), all_data[:, 2].max())
    return x_limits, y_limits, z_limits

# 欠陥なしデータと層ごとのデータをPCAに変換する関数
def safe_pca_transform(data):
    if len(data) > 0:  # データが空でないかを確認
        data = np.atleast_2d(data)  # 1次元データを2次元に変換
        return pca.transform(data)
    else:
        return np.array([])  # 空のデータの場合、処理をスキップ

# 欠陥なしデータと層ごとのデータをまとめてPCAした後に範囲を計算
all_pca_data = [safe_pca_transform(defect_free_encoded)] + [safe_pca_transform(encoded_data) for encoded_data in layer_wise_encoded.values()]
all_pca_data = [data for data in all_pca_data if len(data) > 0]  # 空のデータは削除
x_lim, y_lim, z_lim = get_axis_limits(all_pca_data)

# 層ごとのデータを正しいカラーマップで可視化
def visualize_layer_data_3d():
    fig = plt.figure(figsize=(10, 8))
    ax = fig.add_subplot(111, projection='3d')
    
    colors = plt.get_cmap('tab20', len(layer_wise_encoded))
    
    # 層番号でソートした順にプロット
    sorted_layers = sorted(layer_wise_encoded.items(), key=lambda x: x[0])
    
    # 各層ごとのデータを3次元プロット（透明度を高くして、サイズをさらに小さく）
    for idx, (layer_number, encoded_data) in enumerate(sorted_layers):
        z_pca = safe_pca_transform(encoded_data)
        if z_pca.size > 0:  # データが空でない場合のみプロット
            ax.scatter(z_pca[:, 0], z_pca[:, 1], z_pca[:, 2], 
                       label=f'Layer {layer_number}', color=colors(idx), s=5, alpha=0.05)  # alpha=0.05でさらに薄く表示
    
    ax.set_title('3D Latent Space Visualization by Layer and Defect-Free Data')
    ax.set_xlabel('PCA Component 1')
    ax.set_ylabel('PCA Component 2')
    ax.set_zlabel('PCA Component 3')
    
    # PCAの軸の範囲を設定
    ax.set_xlim(x_lim)
    ax.set_ylim(y_lim)
    ax.set_zlim(z_lim)
    
    # 凡例の順番を層番号で統一
    plt.legend(loc='upper right')
    plt.show()

# 層ごとの3次元可視化
visualize_layer_data_3d()

# クラスタリングを実行（K-means）
def perform_kmeans_clustering(n_clusters=2):
    # 全てのデータをまとめてクラスタリング
    all_encoded_data = np.vstack(all_pca_data)
    
    # K-meansでクラスタリング
    kmeans = KMeans(n_clusters=n_clusters, random_state=42)
    labels = kmeans.fit_predict(all_encoded_data)
    
    # クラスタリング結果を可視化
    fig = plt.figure(figsize=(10, 8))
    ax = fig.add_subplot(111, projection='3d')
    
    scatter = ax.scatter(all_encoded_data[:, 0], all_encoded_data[:, 1], all_encoded_data[:, 2], 
                         c=labels, cmap='viridis', s=5, alpha=0.05)
    
    ax.set_title('3D K-means Clustering of Latent Space')
    ax.set_xlabel('PCA Component 1')
    ax.set_ylabel('PCA Component 2')
    ax.set_zlabel('PCA Component 3')
    
    plt.colorbar(scatter)
    plt.show()

# クラスタリングの実行
perform_kmeans_clustering()

# DBSCANでのクラスタリングも試すことができます。
```

```python
import numpy as np
from mpl_toolkits.mplot3d import Axes3D
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

# ダウンサンプリングの関数
def downsample_data(data, sample_size):
    """データをランダムにダウンサンプリングする"""
    if len(data) > sample_size:
        indices = np.random.choice(len(data), size=sample_size, replace=False)
        return data[indices]
    else:
        return data

# PCAで3次元への次元削減
pca = PCA(n_components=3)
pca.fit(all_encoded)  # all_encodedは全てのデータを含む配列

# 寄与率の出力
explained_variance_ratio = pca.explained_variance_ratio_
cumulative_variance_ratio = np.cumsum(explained_variance_ratio)

print("各主成分の寄与率:", explained_variance_ratio)
print("累積寄与率:", cumulative_variance_ratio)

# ダウンサンプリングサイズの指定
sample_size = 5 # 可視化するためのデータサイズを減らす

# 欠陥なしデータと層データを同時に可視化する関数
def visualize_defect_free_and_layer_data_3d():
    defect_free_x, defect_free_y = prepare_data([defect_free_pair])

    fig = plt.figure(figsize=(10, 8))
    ax = fig.add_subplot(111, projection='3d')

    # 欠陥なしデータのプロット
    if defect_free_x is not None and defect_free_y is not None:
        model.eval()
        with torch.no_grad():
            defect_free_encoded = model.encode(defect_free_x, edge_index)
            defect_free_encoded_np = defect_free_encoded.cpu().numpy()
        
        # ダウンサンプリング
        defect_free_encoded_np = downsample_data(defect_free_encoded_np, sample_size)
        
        # 3次元PCAで次元削減
        defect_free_pca = pca.transform(defect_free_encoded_np)
        
        # 欠陥なしデータをプロット（黒色）
        ax.scatter(defect_free_pca[:, 0], defect_free_pca[:, 1], defect_free_pca[:, 2], 
                   color='black', label='NoDefect', s=20, alpha=0.5)

    # L2~10のデータをプロット
    for layer_number in range(2, 11):
        if layer_number in layer_wise_encoded:
            encoded_data = downsample_data(layer_wise_encoded[layer_number], sample_size)
            z_pca = pca.transform(encoded_data)
            ax.scatter(z_pca[:, 0], z_pca[:, 1], z_pca[:, 2], 
                       label=f'Layer {layer_number}', color='orange', s=5, alpha=0.5)

    # L11~19のデータをプロット
    for layer_number in range(11, 20):
        if layer_number in layer_wise_encoded:
            encoded_data = downsample_data(layer_wise_encoded[layer_number], sample_size)
            z_pca = pca.transform(encoded_data)
            ax.scatter(z_pca[:, 0], z_pca[:, 1], z_pca[:, 2], 
                       label=f'Layer {layer_number}', color='red', s=5, alpha=0.5)

    ax.set_title('3D Visualization of Defect-Free and Defect Data (L2~10, L11~19)')
    ax.set_xlabel('PCA Component 1')
    ax.set_ylabel('PCA Component 2')
    ax.set_zlabel('PCA Component 3')
    ax.legend(loc='center left', bbox_to_anchor=(1.05, 0.5), fontsize='small')
    plt.show()

# 可視化の実行
# visualize_defect_free_data_3d()  # 欠陥なしデータの可視化
# visualize_layer_data_3d()  # 欠陥ありデータの可視化
visualize_defect_free_and_layer_data_3d()  # 両方のデータを同時に可視化
```
