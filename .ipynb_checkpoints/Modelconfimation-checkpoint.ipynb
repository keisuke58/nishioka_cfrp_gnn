```python
import os
import torch
import numpy as np
from torch_geometric.data import Data
from torch_geometric.nn import GCNConv
import torch.nn.functional as F
import matplotlib.pyplot as plt

# GCNモデルの定義
class GCNModel(torch.nn.Module):
    def __init__(self):
        super(GCNModel, self).__init__()
        self.conv1 = GCNConv(4, 16)  # 入力は4次元 (x, y, z, 応力)
        self.conv2 = GCNConv(16, 32)
        self.conv3 = GCNConv(32, 16)

        # 出力層を1つ用意（欠陥の程度を予測）
        self.fc_defect = torch.nn.Linear(16, 1)  # 回帰タスクのため1次元出力

    def forward(self, data):
        x, edge_index = data.x, data.edge_index
        print(f"Input shape: {x.shape}")  # 最初の入力
        x = F.relu(self.conv1(x, edge_index))
        print(f"After conv1: {x.shape}")  # conv1の後
        x = F.relu(self.conv2(x, edge_index))
        print(f"After conv2: {x.shape}")  # conv2の後
        x = F.relu(self.conv3(x, edge_index))
        print(f"After conv3: {x.shape}")  # conv3の後
        output = self.fc_defect(x)
        print(f"Output shape: {output.shape}")  # 出力
        return output

# 学習済みモデルのロード
def load_trained_model(model_path):
    model = GCNModel()
    model.load_state_dict(torch.load(model_path))  # 学習済みのパラメータをロード
    model.eval()  # 評価モードに設定
    return model

# モデルのパラメータを確認する
def check_model_parameters(model):
    for name, param in model.named_parameters():
        print(f"Layer: {name}, Shape: {param.shape}, Values: {param[:2]}")  # パラメータの形状と最初の2つの値を表示

# モデルのロード後にパラメータを確認
model = load_trained_model(model_path)
check_model_parameters(model)

# データの準備
def prepare_test_data(file_path, edge_path):
    # 入力データの読み込み (x, y, z 座標と主応力データ)
    x_coords = np.load("/home/nishioka/GNN/BasicdataforGNN/x_2layer.npy")[:3654]
    y_coords = np.load("/home/nishioka/GNN/BasicdataforGNN/y_2layer.npy")[:3654]
    z_coords = np.load("/home/nishioka/GNN/BasicdataforGNN/z_2layer.npy")[:3654]
    stress_values = np.load(file_path)  # 主応力データ

    # エッジ情報の読み込み
    edges = np.load(edge_path)

    # ノード特徴量 (x, y, z, 主応力) を結合
    node_features = np.vstack((x_coords, y_coords, z_coords, stress_values)).T

    # データを PyTorch のテンソルに変換
    input_data = torch.tensor(node_features, dtype=torch.float)
    edge_index = torch.tensor(edges.T, dtype=torch.long)

    # torch_geometric の Data オブジェクトに変換
    data = Data(x=input_data, edge_index=edge_index)
    return data

# 予測を実行し、結果を表示する関数
def predict(model, data):
    model.eval()
    with torch.no_grad():  # 勾配計算を無効にして予測
        output = model(data)
    print(f"Model output (before converting to NumPy): {output[:100]}")  # 先頭10個の予測値を表示
    return output

# 予測結果と教師データの比較
def compare_with_labels(output, label_file_path):
    # 出力と教師データを比較
    true_labels = np.load(label_file_path)[:3654]  # ラベルを読み込む
    output_np = output.detach().cpu().numpy()

    # 予測結果と教師データの最初の10個を比較表示
    print("Predicted vs True Labels (First 10 Samples):")
    for i in range(10):
        print(f"Predicted: {output_np[i][0]:.4f}, True: {true_labels[i]:.4f}")

    # 統計情報を表示
    print(f"\nOutput Stats: min={output_np.min():.4f}, max={output_np.max():.4f}, mean={output_np.mean():.4f}")
    print(f"Label Stats: min={true_labels.min():.4f}, max={true_labels.max():.4f}, mean={true_labels.mean():.4f}")

# 予測結果を表示・プロット
def plot_predictions(output):
    # 予測結果を NumPy 配列に変換
    output_np = output.detach().cpu().numpy()

    # ヒストグラムで予測結果を可視化
    plt.figure(figsize=(10, 6))
    plt.hist(output_np, bins=50, color='blue', alpha=0.7)
    plt.title("Prediction Results Distribution")
    plt.xlabel("Predicted Values")
    plt.ylabel("Frequency")
    plt.show()

# 予測結果と教師データの分布をヒストグラムで可視化
def plot_predictions_and_labels(output, label_file_path):
    # 教師データを読み込み
    true_labels = np.load(label_file_path)[:3654]  # ラベルを読み込む

    # 予測結果をNumPyに変換
    output_np = output.detach().cpu().numpy()

    # 予測値と教師データのヒストグラムを並べて表示
    plt.figure(figsize=(12, 6))

    # 予測結果のヒストグラム
    plt.subplot(1, 2, 1)
    plt.hist(output_np, bins=50, color='blue', alpha=0.7)
    plt.title("Prediction Results Distribution")
    plt.xlabel("Predicted Values")
    plt.ylabel("Frequency")

    # 教師データのヒストグラム
    plt.subplot(1, 2, 2)
    plt.hist(true_labels, bins=50, color='orange', alpha=0.7)
    plt.title("True Labels Distribution")
    plt.xlabel("True Label Values")
    plt.ylabel("Frequency")

    # プロットを表示
    plt.tight_layout()
    plt.show()

# モデルファイルが存在しているか確認
model_path = '/home/nishioka/GNN/GNNmodel.pth/gcn_model_first_testloss0.0001.pth'
if os.path.exists(model_path):
    print(f"Model file found at: {model_path}")
else:
    print(f"Model file not found: {model_path}")

# 読み込んだデータが異なるかを確認する
def prepare_test_data(file_path, edge_path):
    x_coords = np.load("/home/nishioka/GNN/BasicdataforGNN/x_2layer.npy")[:3654]
    y_coords = np.load("/home/nishioka/GNN/BasicdataforGNN/y_2layer.npy")[:3654]
    z_coords = np.load("/home/nishioka/GNN/BasicdataforGNN/z_2layer.npy")[:3654]
    stress_values = np.load(file_path)
    
    # 入力データを表示して確認する
    print(f"Stress values from {file_path}: {stress_values[:10]}")
    
    node_features = np.vstack((x_coords, y_coords, z_coords, stress_values)).T
    input_data = torch.tensor(node_features, dtype=torch.float)
    edge_index = torch.tensor(np.load(edge_path).T, dtype=torch.long)
    
    data = Data(x=input_data, edge_index=edge_index)
    return data

def predict(model, data):
    model.eval()
    with torch.no_grad():  # 勾配計算を無効にして予測
        print(f"Input data shape: {data.x.shape}")
        output = model(data)
        print(f"Output shape: {output.shape}")
    return output
    
# メイン関数
def main():
    # モデルファイルのパス
    model_path = '/home/nishioka/GNN/GNNmodel.pth/gcn_model.pth'
    
    # テストデータのパス
    file_path = "/home/nishioka/GNN/Defect_4x4_Standardized1/Standardized1_Defect4x4_ELNOD_L10B1.npy"
    edge_path = "/home/nishioka/GNN/BasicdataforGNN/edges_2layer.npy"
    label_file_path = "/home/nishioka/GNN/DefectLabels_4x4_test1/DefectLabel_L10B1.npy"  # ラベルファイルのパス

    # モデルのロード
    model = load_trained_model(model_path)

    # テストデータの準備
    data = prepare_test_data(file_path, edge_path)

    # 予測を実行
    output = predict(model, data)

    # 予測結果の表示
    print("予測結果:", output)

    # 予測結果とラベルの比較
    compare_with_labels(output, label_file_path)

    # 予測結果のプロット
    plot_predictions(output)

    plot_predictions_and_labels(output, label_file_path)


if __name__ == "__main__":
    main()
```

```python
import os
import torch
import numpy as np
from torch_geometric.data import Data
from torch_geometric.nn import GCNConv
import torch.nn.functional as F
import matplotlib.pyplot as plt

hidden_channels = 32


# GCNモデルの定義
class GCNModel(torch.nn.Module):
    def __init__(self, hidden_channels=32):
        super(GCNModel, self).__init__()
        self.conv1 = GCNConv(4, hidden_channels)  # 入力は4次元 (x, y, z, 応力)
        self.conv2 = GCNConv(hidden_channels, hidden_channels*2)
        self.conv3 = GCNConv(hidden_channels*2, hidden_channels)

        # 出力層を1つ用意（欠陥の程度を予測）
        self.fc_defect = torch.nn.Linear(hidden_channels, 1)  # 回帰タスクのため1次元出力

    def forward(self, data):
        x, edge_index = data.x, data.edge_index
        print(f"Input shape: {x.shape}")  # 最初の入力
        x = F.relu(self.conv1(x, edge_index))
        print(f"After conv1: {x.shape}")  # conv1の後
        x = F.relu(self.conv2(x, edge_index))
        print(f"After conv2: {x.shape}")  # conv2の後
        x = F.relu(self.conv3(x, edge_index))
        print(f"After conv3: {x.shape}")  # conv3の後
        output = self.fc_defect(x)
        print(f"Output shape: {output.shape}")  # 出力
        return output

# 学習済みモデルのロード
def load_trained_model(model_path):
    model = GCNModel(hidden_channels=hidden_channels)
    model.load_state_dict(torch.load(model_path))  # 学習済みのパラメータをロード
    model.eval()  # 評価モードに設定
    return model

# モデルのパラメータを確認する
def check_model_parameters(model):
    for name, param in model.named_parameters():
        print(f"Layer: {name}, Shape: {param.shape}, Values: {param[:2]}")  # パラメータの形状と最初の2つの値を表示

# モデルのロード後にパラメータを確認
model = load_trained_model(model_path)
check_model_parameters(model)

# データの準備
def prepare_test_data(file_path, edge_path):
    # 入力データの読み込み (x, y, z 座標と主応力データ)
    x_coords = np.load("/home/nishioka/GNN/BasicdataforGNN/x_2layer.npy")[:3654]
    y_coords = np.load("/home/nishioka/GNN/BasicdataforGNN/y_2layer.npy")[:3654]
    z_coords = np.load("/home/nishioka/GNN/BasicdataforGNN/z_2layer.npy")[:3654]
    stress_values = np.load(file_path)  # 主応力データ

    # エッジ情報の読み込み
    edges = np.load(edge_path)

    # ノード特徴量 (x, y, z, 主応力) を結合
    node_features = np.vstack((x_coords, y_coords, z_coords, stress_values)).T

    # データを PyTorch のテンソルに変換
    input_data = torch.tensor(node_features, dtype=torch.float)
    edge_index = torch.tensor(edges.T, dtype=torch.long)

    # torch_geometric の Data オブジェクトに変換
    data = Data(x=input_data, edge_index=edge_index)
    return data

# 予測を実行し、結果を表示する関数
def predict(model, data):
    model.eval()
    with torch.no_grad():  # 勾配計算を無効にして予測
        output = model(data)
    print(f"Model output (before converting to NumPy): {output[:100]}")  # 先頭10個の予測値を表示
    return output

# 予測結果と教師データの比較
def compare_with_labels(output, label_file_path):
    # 出力と教師データを比較
    true_labels = np.load(label_file_path)[:3654]  # ラベルを読み込む
    output_np = output.detach().cpu().numpy()

    # 予測結果と教師データの最初の10個を比較表示
    print("Predicted vs True Labels (First 10 Samples):")
    for i in range(10):
        print(f"Predicted: {output_np[i][0]:.4f}, True: {true_labels[i]:.4f}")

    # 統計情報を表示
    print(f"\nOutput Stats: min={output_np.min():.4f}, max={output_np.max():.4f}, mean={output_np.mean():.4f}")
    print(f"Label Stats: min={true_labels.min():.4f}, max={true_labels.max():.4f}, mean={true_labels.mean():.4f}")

# 予測結果を表示・プロット
def plot_predictions(output):
    # 予測結果を NumPy 配列に変換
    output_np = output.detach().cpu().numpy()

    # ヒストグラムで予測結果を可視化
    plt.figure(figsize=(10, 6))
    plt.hist(output_np, bins=50, color='blue', alpha=0.7)
    plt.title("Prediction Results Distribution")
    plt.xlabel("Predicted Values")
    plt.ylabel("Frequency")
    plt.show()

# 予測結果と教師データの分布をヒストグラムで可視化
def plot_predictions_and_labels(output, label_file_path):
    # 教師データを読み込み
    true_labels = np.load(label_file_path)[:3654]  # ラベルを読み込む

    # 予測結果をNumPyに変換
    output_np = output.detach().cpu().numpy()

    # 予測値と教師データのヒストグラムを並べて表示
    plt.figure(figsize=(12, 6))

    # 予測結果のヒストグラム
    plt.subplot(1, 2, 1)
    plt.hist(output_np, bins=50, color='blue', alpha=0.7)
    plt.title("Prediction Results Distribution")
    plt.xlabel("Predicted Values")
    plt.ylabel("Frequency")

    # 教師データのヒストグラム
    plt.subplot(1, 2, 2)
    plt.hist(true_labels, bins=50, color='orange', alpha=0.7)
    plt.title("True Labels Distribution")
    plt.xlabel("True Label Values")
    plt.ylabel("Frequency")

    # プロットを表示
    plt.tight_layout()
    plt.show()

# モデルファイルが存在しているか確認
model_path = '/home/nishioka/GNN/GNNmodel.pth/gcn_model_first_testloss0.0001.pth'
if os.path.exists(model_path):
    print(f"Model file found at: {model_path}")
else:
    print(f"Model file not found: {model_path}")

# 読み込んだデータが異なるかを確認する
def prepare_test_data(file_path, edge_path):
    x_coords = np.load("/home/nishioka/GNN/BasicdataforGNN/x_2layer.npy")[:3654]
    y_coords = np.load("/home/nishioka/GNN/BasicdataforGNN/y_2layer.npy")[:3654]
    z_coords = np.load("/home/nishioka/GNN/BasicdataforGNN/z_2layer.npy")[:3654]
    stress_values = np.load(file_path)
    
    # 入力データを表示して確認する
    print(f"Stress values from {file_path}: {stress_values[:10]}")
    
    node_features = np.vstack((x_coords, y_coords, z_coords, stress_values)).T
    input_data = torch.tensor(node_features, dtype=torch.float)
    edge_index = torch.tensor(np.load(edge_path).T, dtype=torch.long)
    
    data = Data(x=input_data, edge_index=edge_index)
    return data

def predict(model, data):
    model.eval()
    with torch.no_grad():  # 勾配計算を無効にして予測
        print(f"Input data shape: {data.x.shape}")
        output = model(data)
        print(f"Output shape: {output.shape}")
    return output
    
# メイン関数
def main():
    # モデルファイルのパス
    model_path = '/home/nishioka/GNN/gcn_model_0.0084.pth'
    
    # テストデータのパス
    file_path = "/home/nishioka/GNN/Defect_4x4_Standardized1/Standardized1_Defect4x4_ELNOD_L10B1.npy"
    edge_path = "/home/nishioka/GNN/BasicdataforGNN/edges_2layer.npy"
    label_file_path = "/home/nishioka/GNN/DefectLabels_4x4_test1/DefectLabel_L10B1.npy"  # ラベルファイルのパス

    # モデルのロード
    model = load_trained_model(model_path)

    # テストデータの準備
    data = prepare_test_data(file_path, edge_path)

    # 予測を実行
    output = predict(model, data)

    # 予測結果の表示
    print("予測結果:", output)

    # 予測結果とラベルの比較
    compare_with_labels(output, label_file_path)

    # 予測結果のプロット
    plot_predictions(output)

    plot_predictions_and_labels(output, label_file_path)


if __name__ == "__main__":
    main()
```
