```python
import numpy as np
import os

def generate_defect_vertices(start_vertex, num_blocks):
    """
    各ブロックの開始頂点番号を生成する関数。
    
    Parameters:
    start_vertex (int): 最初の頂点番号（例：32）。
    num_blocks (int): 生成するブロックの数（例：90）。
    
    Returns:
    list: 各ブロックの開始頂点番号のリスト。
    """
    defect_vertices = []

    # 各ブロックの開始頂点番号を生成
    for block in range(1, num_blocks + 1):
        vertices = [
            int(start_vertex), int(start_vertex) + 4, int(start_vertex) + 8, int(start_vertex) + 12, int(start_vertex) + 16, int(start_vertex) + 20,
            int(start_vertex) + 116, int(start_vertex) + 120, int(start_vertex) + 124, int(start_vertex) + 128, int(start_vertex) + 132, int(start_vertex) + 136,
            int(start_vertex) + 232, int(start_vertex) + 236, int(start_vertex) + 240, int(start_vertex) + 244, int(start_vertex) + 248, int(start_vertex) + 252,
            int(start_vertex) + 348, int(start_vertex) + 352, int(start_vertex) + 356, int(start_vertex) + 360, int(start_vertex) + 364, int(start_vertex) + 368,
            int(start_vertex) + 464, int(start_vertex) + 468, int(start_vertex) + 472, int(start_vertex) + 476, int(start_vertex) + 480, int(start_vertex) + 484,
            int(start_vertex) + 580, int(start_vertex) + 584, int(start_vertex) + 588, int(start_vertex) + 592, int(start_vertex) + 596, int(start_vertex) + 600,
            int(start_vertex) + 696, int(start_vertex) + 700, int(start_vertex) + 704, int(start_vertex) + 708, int(start_vertex) + 712, int(start_vertex) + 716,
            int(start_vertex) + 812, int(start_vertex) + 816, int(start_vertex) + 820, int(start_vertex) + 824, int(start_vertex) + 828, int(start_vertex) + 832,
            int(start_vertex) + 928, int(start_vertex) + 932, int(start_vertex) + 936, int(start_vertex) + 940, int(start_vertex) + 944, int(start_vertex) + 948,
            int(start_vertex) + 1044, int(start_vertex) + 1048, int(start_vertex) + 1052, int(start_vertex) + 1056, int(start_vertex) + 1060, int(start_vertex) + 1064,
            int(start_vertex) + 1160, int(start_vertex) + 1164, int(start_vertex) + 1168, int(start_vertex) + 1172, int(start_vertex) + 1176, int(start_vertex) + 1180,
            int(start_vertex) + 1276, int(start_vertex) + 1280, int(start_vertex) + 1284, int(start_vertex) + 1288, int(start_vertex) + 1292, int(start_vertex) + 1296,
            int(start_vertex) + 1392, int(start_vertex) + 1396, int(start_vertex) + 1400, int(start_vertex) + 1404, int(start_vertex) + 1408, int(start_vertex) + 1412,
            int(start_vertex) + 1508, int(start_vertex) + 1512, int(start_vertex) + 1516, int(start_vertex) + 1520, int(start_vertex) + 1524, int(start_vertex) + 1528,
            int(start_vertex) + 1624, int(start_vertex) + 1628, int(start_vertex) + 1632, int(start_vertex) + 1636, int(start_vertex) + 1640, int(start_vertex) + 1644
        ]
        # defect_vertices.append(vertices)  # append the vertices for each block

    return vertices



def generate_blocks(start_vertices, num_rows=29):
    """
    ブロック内の頂点番号を生成する関数。
    
    Parameters:
    start_vertices (list): 各ブロックの開始頂点番号のリスト。
    num_rows (int): 縦方向の行数（デフォルトは29行）。
    
    Returns:
    blocks (list): 各ブロックに対応する頂点番号のリストを格納したリスト。
    """
    blocks = []
    
    for start in start_vertices:
        block = [
            start, start + 1, start + 2, start + 3, start + 4,
            start + num_rows, start + num_rows + 1, start + num_rows + 2, start + num_rows + 3, start + num_rows + 4,
            start + num_rows * 2, start + num_rows * 2 + 1, start + num_rows * 2 + 2, start + num_rows * 2 + 3, start + num_rows * 2 + 4,
            start + num_rows * 3, start + num_rows * 3 + 1, start + num_rows * 3 + 2, start + num_rows * 3 + 3, start + num_rows * 3 + 4,
            start + num_rows * 4, start + num_rows * 4 + 1, start + num_rows * 4 + 2, start + num_rows * 4 + 3, start + num_rows * 4 + 4
        ]
        blocks.append(block)
    
    print(f"Generated {len(blocks)} blocks. Here are the first few blocks:")
    for i, block in enumerate(blocks[:5]):  # 最初の5つのブロックを表示
        print(f"Block {i+1}: {block[:5]}")  # 各ブロックの最初の5つの頂点を表示
    
    return blocks

    
def assign_labels_to_blocks(block1_vertices,block2_vertices, layer, num_classes=19):
    labels = np.zeros((3654, num_classes))  # 各頂点に19クラスのOne-Hotラベルを初期化

    # 欠陥なしの場合のラベルを [1, 0, 0, 0, ..., 0] とする
    no_defect_label = np.array([1] + [0] * (num_classes - 1))
    
    # 欠陥ありの場合のラベルを層に応じて設定
    defect_label = np.zeros(num_classes)
    defect_label[0] = 0  # 初期値として [1, 0, 0, 0, ..., 0]

    # 各層に対応するOne-Hotラベルを設定（2〜19層）
    if 2 <= layer <= 19:
        defect_label[layer - 1] = 1  # 例: 層2なら [0, 1, 0, 0, ..., 0]

    # 全頂点に初期ラベル（欠陥なし）を設定
    labels[:] = no_defect_label

    # 欠陥のある頂点に対してラベルを上書き
    if 2 <= layer <= 10:
        # Layer 2~10 は block1_vertices に欠陥ラベルを設定
        for vertex in block1_vertices:
            labels[vertex] = defect_label
        print(f"Layer {layer}: block1_vertices = {block1_vertices[:5]}")  # デバッグ用出力
    elif 11 <= layer <= 19:
        # Layer 11~19 は block2_vertices に欠陥ラベルを設定
        for vertex in block2_vertices:
            labels[vertex] = defect_label
        print(f"Layer {layer}: block2_vertices = {block2_vertices[:5]}")  # デバッグ用出力

    # # 欠陥のある頂点に対してラベルを上書き
    # for vertex in block1_vertices:
    #     if vertex < 1827:
    #         labels[vertex] = defect_label

    # print(f"Layer {layer}: block1_vertices = {block1_vertices[:5]}")  # デバッグ用出力

    # for vertex in block2_vertices:
    #     if vertex > 1827:
    #         labels[vertex] = defect_label

    # print(f"Layer {layer}: block2_vertices = {block2_vertices[:5]}")  # デバッグ用出力
            
    return labels


def generate_and_save_defect_labels(start_vertex1, start_vertex2, num_blocks, num_layers, output_folder):
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)

    # 各ブロックの頂点を生成
    start_vertices1 = generate_defect_vertices(start_vertex1, num_blocks)
    blocks1 = generate_blocks(start_vertices1)

    start_vertices2 = generate_defect_vertices(start_vertex2, num_blocks)
    blocks2 = generate_blocks(start_vertices2)

    # 各層ごとにラベルを割り当ててファイルに保存
    for layer in range(2, num_layers + 1):  # 層は2〜19
        for i in range(num_blocks):
            labels = assign_labels_to_blocks(blocks1[i], blocks2[i], layer)  # layerは整数として渡す
            filename = os.path.join(output_folder, f"Defect19Class_L{layer}B{i+1}.npy")
            np.save(filename, labels)
            print(f"Saved file: {filename}")


# 実行例
start_vertex1 = 32
start_vertex2 = 1859
num_blocks = 90
num_layers = 19
output_folder = "Defect19Class_OneHot_test3"

generate_and_save_defect_labels(start_vertex1, start_vertex2, num_blocks, num_layers, output_folder)
```
