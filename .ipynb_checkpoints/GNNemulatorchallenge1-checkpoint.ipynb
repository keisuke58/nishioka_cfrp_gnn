```python
import jax.numpy as jnp
from flax import linen as nn
from typing import Sequence
import jax.numpy as jnp
import optax



# トレーニングデータ
V_train = jnp.zeros((3654000, 4))  # 例として全て0で初期化（実データに置き換えてください）
senders_train = jnp.array([0, 1, 2, ...])  # 送信ノードのインデックス
receivers_train = jnp.array([1, 2, 0, ...])  # 受信ノードのインデックス

# 検証データ
V_val = jnp.zeros((1132740, 4))  # 検証用のノード特徴
senders_val = jnp.array([0, 1, 2, ...])
receivers_val = jnp.array([1, 2, 0, ...])

# テストデータ
V_test = jnp.zeros((1136394, 4))  # テスト用のノード特徴
senders_test = jnp.array([0, 1, 2, ...])
receivers_test = jnp.array([1, 2, 0, ...])

# グローバルパラメータ（theta）
theta = jnp.array([0.5, 0.2, 0.8])  # グローバルなパラメータは適切に設定


# DeepGraphEmulatorモデル
class DeepGraphEmulator(nn.Module):
    mlp_features: Sequence[int]
    latent_size: Sequence[int]
    K: int
    receivers: Sequence[int]
    senders: Sequence[int]
    n_total_nodes: int
    output_dim: int
    real_node_indices: Sequence[bool]

    @nn.compact
    def __call__(self, V: jnp.ndarray, E: jnp.ndarray, theta: jnp.ndarray, z_global: jnp.ndarray = None, sow_latents: bool = False):
        # エンコーダー
        node_encode_mlp = make_layernorm_mlp(self.mlp_features + self.latent_size)
        edge_encode_mlp = make_layernorm_mlp(self.mlp_features + self.latent_size)
        
        # メッセージパッシングステップ
        message_passing_blocks = [MessagePassingStep(make_layernorm_mlp(self.mlp_features + self.latent_size),
                                                     make_layernorm_mlp(self.mlp_features + self.latent_size),
                                                     self.senders, self.receivers, self.n_total_nodes) for i in range(self.K)]
        
        # グローバルパラメータのエンコード
        theta_encode_mlp = make_layernorm_mlp(self.mlp_features + self.latent_size)
        
        # デコーダー
        node_decode_mlps = [make_mlp(self.mlp_features + (1,)) for i in range(self.output_dim[0])]

        # ノードとエッジのエンコード
        V = node_encode_mlp(V)
        E = edge_encode_mlp(E)
        
        # メッセージパッシング
        for message_pass_block_i in message_passing_blocks:
            V, E = message_pass_block_i(V, E)
        
        # メッセージの集約
        incoming_messages = aggregate_incoming_messages(E, self.receivers, self.n_total_nodes)
        z_local = jnp.hstack((V, incoming_messages))
        z_local = z_local[self.real_node_indices]
        
        if sow_latents:
            return z_local

        # グローバルパラメータの処理
        z_theta = theta_encode_mlp(theta)
        globals_array = jnp.tile(z_theta, (z_local.shape[0], 1)) if z_global is None else jnp.tile(jnp.hstack((z_theta, z_global.reshape(1, -1))), (z_local.shape[0], 1))
        
        # ノードの最終的な表現
        final_representation = jnp.hstack((globals_array, z_local))
        individual_mlp_predictions = [decode_mlp(final_representation) for decode_mlp in node_decode_mlps]
        
        # 変位予測
        Upred = jnp.hstack(individual_mlp_predictions)
        return Upred

model = DeepGraphEmulator(
    mlp_features=[64, 128],  # MLP層のノード数
    latent_size=[256],  # 潜在ベクトルのサイズ
    K=3,  # メッセージパッシングのステップ数
    receivers=receivers_train,
    senders=senders_train,
    n_total_nodes=3654000,  # トレーニングデータのノード数
    output_dim=(1,),  # 出力次元 (変位の予測)
    real_node_indices=jnp.array([True for _ in range(3654000)])  # ノードが使用されるか
)

# ノード特徴行列 (V_train), エッジ特徴行列 (E_train), グローバルパラメータ (theta) を使用して予測を行う
predictions_train = model.apply({'params': model.init(jax.random.PRNGKey(0), V_train, E_train, theta)}, V_train, E_train, theta)
print(predictions_train)

# MSE損失関数の定義
def loss_fn(params, V, E, theta, true_displacements):
    predictions = model.apply({'params': params}, V, E, theta)
    return jnp.mean((predictions - true_displacements) ** 2)

# オプティマイザーを定義
optimizer = optax.adam(learning_rate=0.001)
opt_state = optimizer.init(model)

# 学習ステップの実行
grads = jax.grad(loss_fn)(model.params, V_train, E_train, theta, true_displacements_train)
updates, opt_state = optimizer.update(grads, opt_state, model.params)
model = optax.apply_updates(model, updates)

```
