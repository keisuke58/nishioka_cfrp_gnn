```python
import torch
from torch_geometric.data import Data
import numpy as np

# VTKファイルのデータをロード（ここでは既存のデータを使用）
x = np.load("BasicdataforGNN/x_2layer.npy")[:3654]  # ノードの座標データ (例)
y = np.load("BasicdataforGNN/y_2layer.npy")[:3654]
z = np.load("BasicdataforGNN/z_2layer.npy")[:3654]
edges = np.load("BasicdataforGNN/edges_2layer.npy")[:3654]  # エッジ情報
values = np.load("Defect_4x4_Standardized1/0Standardized1_nodefect_4x4_ElNOD.npy")  # 応力値
# values = np.load("Defect_4x4_Standardized1/Standardized1_Defect4x4_ELNOD_L10B1.npy")[:3654, 1]  # 応力値

# ノード特徴量：座標情報と応力値を組み合わせる
node_features = np.vstack((x, y, z, values)).T  # 座標+応力値 (shape: [N, 4])
print(node_features.shape)

# エッジリスト：PyTorch Geometricのフォーマットに変換
edge_index = torch.tensor(edges.T, dtype=torch.long)  # [2, E] 形式でエッジリストを作成
print(edge_index.shape)

# ノード特徴量をテンソルに変換
x = torch.tensor(node_features, dtype=torch.float)  # [N, 4] 形式でノード特徴量
print(x.shape)

# 欠陥位置のラベル（ここでは仮にランダムに生成。実データに基づいたラベルが必要）
# labels = np.load("DefectLabels_4x4_test1/DefectLabel_L10B12.npy")[:3654, 1]
labels = torch.tensor(np.random.randint(0, 2, size=len(x)), dtype=torch.long)  # 0/1のラベル
print(labels.shape)

# データをPyTorch GeometricのDataクラスにまとめる
data = Data(x=x, edge_index=edge_index, y=labels)
print(data.shape)
```

```python
import torch
import torch.nn.functional as F
from torch_geometric.nn import GCNConv

class GCN(torch.nn.Module):
    def __init__(self, in_channels, hidden_channels, out_channels):
        super(GCN, self).__init__()
        # 1層目のGCN層
        self.conv1 = GCNConv(in_channels, hidden_channels)
        # 2層目のGCN層
        self.conv2 = GCNConv(hidden_channels, out_channels)

    def forward(self, data):
        x, edge_index = data.x, data.edge_index

        # 1層目のGCNとReLU活性化関数
        x = self.conv1(x, edge_index)
        x = F.relu(x)

        # 2層目のGCNとソフトマックス
        x = self.conv2(x, edge_index)
        return F.log_softmax(x, dim=1)

# モデルの定義
model = GCN(in_channels=4, hidden_channels=16, out_channels=2)  # 入力4次元（座標+応力値）、2クラス（欠陥/非欠陥）
```

```python
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)
criterion = torch.nn.CrossEntropyLoss()

# トレーニングのループ
def train():
    model.train()
    optimizer.zero_grad()
    out = model(data)
    loss = criterion(out, data.y)
    loss.backward()
    optimizer.step()
    return loss.item()

# エポックごとのトレーニング
for epoch in range(100):
    loss = train()
    if epoch % 10 == 0:
        print(f'Epoch {epoch}, Loss: {loss}')
```

```python
model.eval()
with torch.no_grad():
    out = model(data)
    pred = out.argmax(dim=1)  # 欠陥位置の予測
    correct = (pred == data.y).sum()
    acc = int(correct) / len(data.y)
    print(f'Accuracy: {acc:.4f}')
```

```python
import torch
import torch.nn.functional as F
from torch_geometric.nn import GCNConv
from torch_geometric.data import Data
import numpy as np

# Randomly generate values (with no-defect cases included)
num_nodes = 1000  # Number of nodes to simulate
x = np.random.rand(num_nodes)  # Random x-coordinates
y = np.random.rand(num_nodes)  # Random y-coordinates
z = np.random.rand(num_nodes)  # Random z-coordinates
values = np.random.rand(num_nodes)  # Random stress values

# Ensure at least one no-defect case
values[0] = 0  # Ensure at least one node has no stress (representing no defect)

# Randomly generate labels (0 for no defect, 1 for defect)
labels = np.random.randint(0, 2, size=num_nodes)  # Random labels for defect/no-defect
labels[0] = 0  # Ensure that the no-defect case has a label of 0

# Simulate some random edge connections between nodes (e.g., 2 edges per node)
num_edges = num_nodes * 2  # Number of edges (for simplicity, assume 2 edges per node)
edges = np.random.randint(0, num_nodes, size=(2, num_edges))  # Random edge connections

# Combine node features: [x, y, z, stress values]
node_features = np.vstack((x, y, z, values)).T  # Shape: [N, 4]

# Convert to PyTorch tensors
x = torch.tensor(node_features, dtype=torch.float)  # Node features [N, 4]
edge_index = torch.tensor(edges, dtype=torch.long)  # Edge list [2, E]
labels = torch.tensor(labels, dtype=torch.long)  # Labels [N]

# Create PyTorch Geometric Data object
data = Data(x=x, edge_index=edge_index, y=labels)

# Define a simple GCN model
class GCN(torch.nn.Module):
    def __init__(self, input_dim, hidden_dim, output_dim):
        super(GCN, self).__init__()
        self.conv1 = GCNConv(input_dim, hidden_dim)  # First graph convolution
        self.conv2 = GCNConv(hidden_dim, output_dim)  # Second graph convolution

    def forward(self, data):
        x, edge_index = data.x, data.edge_index

        # First graph convolution + ReLU
        x = self.conv1(x, edge_index)
        x = F.relu(x)

        # Second graph convolution (output layer)
        x = self.conv2(x, edge_index)

        return F.log_softmax(x, dim=1)  # Use log softmax for classification

# Initialize GCN model
input_dim = 4  # Node features: [x, y, z, stress values]
hidden_dim = 16  # Number of hidden units
output_dim = 2  # 2 classes: defect (1), no defect (0)
model = GCN(input_dim, hidden_dim, output_dim)

# Move the model and data to CUDA if available
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
model = model.to(device)
data = data.to(device)

# Forward pass through the model
output = model(data)

# Print the output predictions for the first 5 nodes
print("Predictions for the first 5 nodes:\n", output[:5])

# Calculate the accuracy (optional, for example)
pred = output.argmax(dim=1)  # Get predicted class (0 or 1)
correct = (pred == data.y).sum().item()  # Count correct predictions
accuracy = correct / num_nodes  # Calculate accuracy
print(f"Accuracy: {accuracy * 100:.2f}%")
```

```python
import torch
import torch.nn.functional as F
from torch_geometric.nn import GCNConv
from torch_geometric.data import Data
from torch_geometric.loader import DataLoader
import numpy as np

# 実際のデータをロード
x = np.load("BasicdataforGNN/x_2layer.npy")[:3654]  # ノードのx座標データ
y = np.load("BasicdataforGNN/y_2layer.npy")[:3654]  # ノードのy座標データ
z = np.load("BasicdataforGNN/z_2layer.npy")[:3654]  # ノードのz座標データ
edges = np.load("BasicdataforGNN/edges_2layer.npy")  # エッジ情報
values = np.load("Defect_4x4_Standardized1/0Standardized1_nodefect_4x4_ElNOD.npy")[:3654]  # 応力値
labels = np.load("DefectLabels_4x4_test1/DefectLabel_L10B12.npy")[:3654]  # 欠陥ラベル

# ノード特徴量：座標情報と応力値を組み合わせる
node_features = np.vstack((x, y, z, values)).T  # 座標+応力値 (shape: [N, 4])

# エッジリスト：PyTorch Geometricのフォーマットに変換
edge_index = torch.tensor(edges.T, dtype=torch.long)  # [2, E] 形式でエッジリストを作成

# ノード特徴量とラベルをPyTorchテンソルに変換
x = torch.tensor(node_features, dtype=torch.float)  # [N, 4] 形式でノード特徴量
labels = torch.tensor(labels, dtype=torch.long)  # 欠陥ラベル（0/1）

# データをPyTorch GeometricのDataクラスにまとめる
data = Data(x=x, edge_index=edge_index, y=labels)

# データを訓練データとテストデータに分割
num_train = int(len(data.y) * 0.6)  # 60%を訓練データに
train_data = Data(x=data.x[:num_train], edge_index=data.edge_index[:, :num_train], y=data.y[:num_train])
test_data = Data(x=data.x[num_train:], edge_index=data.edge_index[:, num_train:], y=data.y[num_train:])

# データローダーを作成
train_loader = DataLoader([train_data], batch_size=32, shuffle=True)
test_loader = DataLoader([test_data], batch_size=32)

# GNNモデルの定義
class GCNModel(torch.nn.Module):
    def __init__(self):
        super(GCNModel, self).__init__()
        self.conv1 = GCNConv(4, 16)  # 入力4次元（x, y, z, 応力値）、出力16次元
        self.conv2 = GCNConv(16, 2)  # 出力は2次元（欠陥/欠陥なしのクラス）

    def forward(self, data):
        x, edge_index = data.x, data.edge_index
        x = F.relu(self.conv1(x, edge_index))  # 最初のグラフ畳み込み層とReLU活性化関数
        x = self.conv2(x, edge_index)  # 2番目のグラフ畳み込み層
        return F.log_softmax(x, dim=1)  # 出力をsoftmaxで分類

# モデルのインスタンス化と最適化手法の設定
model = GCNModel()
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)

# モデルの学習
def train():
    model.train()
    total_loss = 0
    for batch in train_loader:
        optimizer.zero_grad()
        out = model(batch)
        loss = F.nll_loss(out, batch.y)
        loss.backward()
        optimizer.step()
        total_loss += loss.item()
    return total_loss / len(train_loader)

# テストデータで精度を評価
def test():
    model.eval()
    correct = 0
    for batch in test_loader:
        out = model(batch)
        pred = out.argmax(dim=1)  # 最も高い値を持つクラスを予測
        correct += (pred == batch.y).sum().item()
    return correct / len(test_data.y)

# エポックごとに学習とテストを繰り返す
for epoch in range(1, 101):  # 100エポックで学習
    loss = train()
    accuracy = test()
    print(f'Epoch {epoch}, Loss: {loss:.4f}, Accuracy: {accuracy * 100:.2f}%')
```

```python
import os
import numpy as np
import torch
from torch_geometric.data import Data

# データフォルダのパス
standardized_data_folder = "Defect_4x4_Standardized1"
label_data_folder = "DefectLabels_4x4_test1"

# データファイルとラベルファイルのリストを取得
data_files = [f for f in os.listdir(standardized_data_folder) if f.startswith("Standardized1_Defect4x4_ELNOD")]
label_files = [f for f in os.listdir(label_data_folder) if f.startswith("DefectLabel_L")]

# 欠陥なしのファイルを追加
data_files.append("0Standardized1_nodefect_4x4_ElNOD.npy")
label_files.append("0DefectLabel_nodefect1.npy")

# L{i}B{j}に対応するデータペアを作成
def extract_layer_block(file_name):
    # 欠陥なしファイルの特別な処理
    if file_name.startswith("0"):
        return (0, 0)  # 欠陥なしデータの場合

    # Layer と Block の情報を抽出
    try:
        # 'L' と 'B' の位置を使って分割し、数値を抽出
        layer_block_str = file_name.split("_")[-1].replace(".npy", "")
        layer_str = layer_block_str.split("L")[1].split("B")[0]  # 'L' の後
        block_str = layer_block_str.split("B")[1]  # 'B' の後

        # Layer と Block を整数に変換
        layer = int(layer_str)
        block = int(block_str)
        return (layer, block)
    except (ValueError, IndexError) as e:
        print(f"無効なファイル名の形式: {file_name}")
        return None  # 無効なファイルはスキップ


# データとラベルファイルを対応付け
data_label_pairs = {}
for data_file in data_files:
    layer_block = extract_layer_block(data_file)
    if layer_block:  # Noneでない場合のみ追加
        data_label_pairs[layer_block] = {"data": data_file}

for label_file in label_files:
    layer_block = extract_layer_block(label_file)
    if layer_block and layer_block in data_label_pairs:  # Noneでない場合のみ追加
        data_label_pairs[layer_block]["label"] = label_file

# 有効なペアのみ取得
valid_pairs = [(v["data"], v["label"]) for k, v in data_label_pairs.items() if "label" in v]

# ランダムに1000個のペアをサンプリング
num_samples = 1000
random_indices = np.random.choice(len(valid_pairs), num_samples, replace=False)

# 座標データのロード
x_coords = np.load("BasicdataforGNN/x_2layer.npy")[:3654]  # ノードのx座標
y_coords = np.load("BasicdataforGNN/y_2layer.npy")[:3654]  # ノードのy座標
z_coords = np.load("BasicdataforGNN/z_2layer.npy")[:3654]  # ノードのz座標

# サンプリングされたデータとラベルを保存するリスト
sampled_data = []
sampled_labels = []
sampled_file_info = []

# サンプルを順次読み込む
for idx in random_indices:
    data_file, label_file = valid_pairs[idx]
    data_file_path = os.path.join(standardized_data_folder, data_file)
    label_file_path = os.path.join(label_data_folder, label_file)
    
    # データとラベルを読み込む
    values = np.load(data_file_path)[:3654]  # 応力値データ
    label = np.load(label_file_path)[:3654]  # 欠陥ラベル
    
    # 座標データと応力データを結合してノード特徴量を作成
    node_features = np.vstack((x_coords, y_coords, z_coords, values)).T  # 座標 + 応力データ
    sampled_data.append(node_features)
    sampled_labels.append(label)
    sampled_file_info.append((data_file, label_file))  # ファイル名のペアを保存

# サンプリングしたデータを結合
sampled_data = np.stack(sampled_data)  # [1000, 3654, 4] となる (サンプル数 x ノード数 x 特徴量数)
sampled_labels = np.hstack(sampled_labels)  # [1000 * 3654] となる

# PyTorchテンソルに変換
x = torch.tensor(sampled_data.reshape(-1, 4), dtype=torch.float)  # 形状を [1000 * 3654, 4] に変換
y = torch.tensor(sampled_labels, dtype=torch.long)  # ラベルも1次元に変換

# エッジ情報を読み込む
edges = np.load("BasicdataforGNN/edges_2layer.npy")
edge_index = torch.tensor(edges.T, dtype=torch.long)

# PyTorch GeometricのDataオブジェクトにまとめる
data = Data(x=x, edge_index=edge_index, y=y)

# サンプリングしたデータの中から5つのペアを出力して確認
print("確認のため、ランダムに選んだ5つのデータとラベルファイルのペアを表示します:")
for i in range(5):
    print(f"データファイル: {sampled_file_info[i][0]}")
    print(f"ラベルファイル: {sampled_file_info[i][1]}")
    print("--------")
```

```python
import os
import numpy as np
import torch
from torch_geometric.data import Data
from sklearn.model_selection import train_test_split

# 座標データのロード
x_coords = np.load("BasicdataforGNN/x_2layer.npy")[:3654]  # ノードのx座標
y_coords = np.load("BasicdataforGNN/y_2layer.npy")[:3654]  # ノードのy座標
z_coords = np.load("BasicdataforGNN/z_2layer.npy")[:3654]  # ノードのz座標

# エッジ情報を読み込む
edges = np.load("BasicdataforGNN/edges_2layer.npy")
edge_index = torch.tensor(edges.T, dtype=torch.long)

# データフォルダ
standardized_data_folder = "Defect_4x4_Standardized1"
label_data_folder = "DefectLabels_4x4_test1"

# データファイルとラベルファイルのリストを取得
data_files = [f for f in os.listdir(standardized_data_folder) if f.startswith("Standardized1_Defect4x4_ELNOD")]
label_files = [f for f in os.listdir(label_data_folder) if f.startswith("DefectLabel_L")]

# 欠陥なしのファイルを追加
data_files.append("0Standardized1_nodefect_4x4_ElNOD.npy")
label_files.append("0DefectLabel_nodefect1.npy")

def extract_layer_block(file_name):
    if file_name.startswith("0"):
        return (0, 0)  # 欠陥なしデータの場合
    try:
        # Layer と Block の情報を抽出
        layer_block_str = file_name.split("_")[-1].replace(".npy", "")
        layer_str = layer_block_str.split("L")[1].split("B")[0]
        block_str = layer_block_str.split("B")[1]
        layer = int(layer_str)
        block = int(block_str)
        return (layer, block)
    except (ValueError, IndexError):
        print(f"無効なファイル名の形式: {file_name}")
        return None  # 無効なファイルはスキップ

# データとラベルファイルを対応付け
data_label_pairs = {}
for data_file in data_files:
    layer_block = extract_layer_block(data_file)
    if layer_block:
        data_label_pairs[layer_block] = {"data": data_file}

for label_file in label_files:
    layer_block = extract_layer_block(label_file)
    if layer_block and layer_block in data_label_pairs:
        data_label_pairs[layer_block]["label"] = label_file

# 有効なペアのみ取得
valid_pairs = [(v["data"], v["label"]) for k, v in data_label_pairs.items() if "label" in v]

# ランダムに1000個のペアをサンプリング
num_samples = 1000
random_indices = np.random.choice(len(valid_pairs), num_samples, replace=False)

# 1000個のサンプルをトレーニングデータとして取得
train_pairs = [valid_pairs[i] for i in random_indices]

# 残りのデータを取得
remaining_pairs = [valid_pairs[i] for i in range(len(valid_pairs)) if i not in random_indices]

# 残りのデータを1:1でvalidationとtestに分割
val_pairs, test_pairs = train_test_split(remaining_pairs, test_size=0.5, random_state=42)

# 検証データとテストデータを適切に分けて格納
train_data = [pair[0] for pair in train_pairs]
train_labels = [pair[1] for pair in train_pairs]

val_data = [pair[0] for pair in val_pairs]
val_labels = [pair[1] for pair in val_pairs]

test_data = [pair[0] for pair in test_pairs]
test_labels = [pair[1] for pair in test_pairs]

# 結果を表示
print(f"Training data size: {len(train_data)}")
print(f"Validation data size: {len(train_labels)}")
print(f"Test data size: {len(test_data)}")
print(f"Training data size: {len(val_labels)}")
print(f"Validation data size: {len(val_data)}")
print(f"Test data size: {len(test_labels)}")

# # トレーニング、バリデーション、テストに分割
# num_train_samples = 1000
# random_indices = np.random.choice(len(valid_pairs), num_train_samples, replace=False)

# 残りのデータを1:1でvalidationとtestに分割
remaining_indices = np.setdiff1d(np.arange(len(valid_pairs)), random_indices)
val_indices, test_indices = train_test_split(remaining_indices, test_size=0.5, random_state=42)

# トレーニング、バリデーション、テスト用のデータリスト
train_pairs = [valid_pairs[i] for i in random_indices]
val_pairs = [valid_pairs[i] for i in val_indices]
test_pairs = [valid_pairs[i] for i in test_indices]

# 関数：サンプルのデータとラベルを作成する
def prepare_data(pairs):
    sampled_data = []
    sampled_labels = []
    
    for data_file, label_file in pairs:
        data_file_path = os.path.join(standardized_data_folder, data_file)
        label_file_path = os.path.join(label_data_folder, label_file)

        # データとラベルを読み込む
        values = np.load(data_file_path)[:3654]  # 応力値データ
        label = np.load(label_file_path)[:3654]  # 欠陥ラベル
        
        # 座標データと応力データを結合してノード特徴量を作成
        node_features = np.vstack((x_coords, y_coords, z_coords, values)).T  # 座標 + 応力データ
        sampled_data.append(node_features)
        sampled_labels.append(label)
    
    # データを結合しテンソルに変換
    sampled_data = np.stack(sampled_data)  # [サンプル数, 3654, 4] となる
    sampled_labels = np.hstack(sampled_labels)  # [サンプル数 * 3654] となる

    x = torch.tensor(sampled_data.reshape(-1, 4), dtype=torch.float)  # 形状を [サンプル数 * 3654, 4] に変換
    y = torch.tensor(sampled_labels, dtype=torch.long)  # ラベルも1次元に変換

    return x, y

# 各データセットを準備
train_x, train_y = prepare_data(train_pairs)
val_x, val_y = prepare_data(val_pairs)
test_x, test_y = prepare_data(test_pairs)

# Dataオブジェクトにまとめる
train_data = Data(x=train_x, edge_index=edge_index, y=train_y)
val_data = Data(x=val_x, edge_index=edge_index, y=val_y)
test_data = Data(x=test_x, edge_index=edge_index, y=test_y)

# 結果を確認
print(f"Training data: {train_data}")
print(f"Validation data: {val_data}")
print(f"Test data: {test_data}")
print(f"Training data: {len(train_data)}")
print(f"Validation data: {len(val_data)}")
print(f"Test data: {len(test_data)}")

# ペアが正しく対応しているか確認
print("\n--- Training データとラベルのペア確認 ---")
for data_file, label_file in train_pairs[:3]:  # トレーニングデータの先頭5個を表示
    print(f"データファイル: {data_file}, ラベルファイル: {label_file}")

print("\n--- Validation データとラベルのペア確認 ---")
for data_file, label_file in val_pairs[:3]:  # 検証データの先頭5個を表示
    print(f"データファイル: {data_file}, ラベルファイル: {label_file}")

print("\n--- Test データとラベルのペア確認 ---")
for data_file, label_file in test_pairs[:3]:  # テストデータの先頭5個を表示
    print(f"データファイル: {data_file}, ラベルファイル: {label_file}")
```

```python
import torch
import torch.nn.functional as F
from torch_geometric.nn import GCNConv
from torch_geometric.data import DataLoader

# 改良されたGCNモデル
class GCNModel(torch.nn.Module):
    def __init__(self):
        super(GCNModel, self).__init__()
        # グラフ畳み込み層
        self.conv1 = GCNConv(4, 16)  # 入力は4次元 (x, y, z, 応力)
        self.conv2 = GCNConv(16, 32) # 中間層
        self.conv3 = GCNConv(32, 16) # 最終層

        # 出力層を3つ用意（欠陥有無、Layer、Block）
        self.fc_defect = torch.nn.Linear(16, 2)  # 欠陥の有無（2クラス）
        self.fc_layer = torch.nn.Linear(16, 18)  # Layerは2~19の18クラス
        self.fc_block = torch.nn.Linear(16, 90)  # Blockは1~90の90クラス

    def forward(self, data):
        x, edge_index = data.x, data.edge_index

        # グラフ畳み込みと活性化関数
        x = F.relu(self.conv1(x, edge_index))
        x = F.relu(self.conv2(x, edge_index))
        x = F.relu(self.conv3(x, edge_index))

        # 出力層
        out1 = self.fc_defect(x)  # 欠陥の有無
        out2 = self.fc_layer(x)   # 欠陥の層
        out3 = self.fc_block(x)   # 欠陥のブロック

        return F.log_softmax(out1, dim=1), F.log_softmax(out2, dim=1), F.log_softmax(out3, dim=1)

# モデル、損失関数、オプティマイザの定義
model = GCNModel()
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)

# 3つのタスクに対応する損失関数
loss_fn_defect = torch.nn.CrossEntropyLoss()
loss_fn_layer = torch.nn.CrossEntropyLoss()
loss_fn_block = torch.nn.CrossEntropyLoss()

# トレーニング関数
def train(loader):
    model.train()
    total_loss = 0
    for batch in loader:
        optimizer.zero_grad()
        out_defect, out_layer, out_block = model(batch)
        
        # 各損失を計算
        loss_defect = loss_fn_defect(out_defect, batch.y_defect)
        loss_layer = loss_fn_layer(out_layer, batch.y_layer)
        loss_block = loss_fn_block(out_block, batch.y_block)
        
        # 総損失
        loss = loss_defect + loss_layer + loss_block
        loss.backward()
        optimizer.step()
        total_loss += loss.item()
    return total_loss / len(loader)

# 検証関数
def test(loader):
    model.eval()
    correct_defect = 0
    correct_layer = 0
    correct_block = 0
    for batch in loader:
        out_defect, out_layer, out_block = model(batch)
        
        # 欠陥の有無
        pred_defect = out_defect.argmax(dim=1)
        correct_defect += (pred_defect == batch.y_defect).sum().item()
        
        # 欠陥の層
        pred_layer = out_layer.argmax(dim=1)
        correct_layer += (pred_layer == batch.y_layer).sum().item()
        
        # 欠陥のブロック
        pred_block = out_block.argmax(dim=1)
        correct_block += (pred_block == batch.y_block).sum().item()
        
    # 正答率の計算
    total_samples = len(loader.dataset)
    accuracy_defect = correct_defect / total_samples
    accuracy_layer = correct_layer / total_samples
    accuracy_block = correct_block / total_samples
    return accuracy_defect, accuracy_layer, accuracy_block

# モデルの学習
for epoch in range(1, 101):
    loss = train(train_loader)
    val_acc_defect, val_acc_layer, val_acc_block = test(val_loader)
    print(f'Epoch {epoch}, Loss: {loss:.4f}, '
          f'Validation Accuracy: Defect {val_acc_defect * 100:.2f}%, '
          f'Layer {val_acc_layer * 100:.2f}%, Block {val_acc_block * 100:.2f}%')

# テストデータでの評価
test_acc_defect, test_acc_layer, test_acc_block = test(test_loader)
print(f'Test Accuracy: Defect {test_acc_defect * 100:.2f}%, '
      f'Layer {test_acc_layer * 100:.2f}%, Block {test_acc_block * 100:.2f}%')
```

```python
import os
import numpy as np
import torch
import torch.nn.functional as F
from torch_geometric.nn import GCNConv
from torch_geometric.data import Data, DataLoader
from sklearn.metrics import mean_squared_error
import pandas as pd
import matplotlib.pyplot as plt
import time

# デバイスの設定 (GPUが使える場合はcudaを使う)
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")

# GCNモデルの定義
class GCNModel(torch.nn.Module):
    def __init__(self, hidden_channels=32):
        super(GCNModel, self).__init__()
        self.conv1 = GCNConv(4, hidden_channels)
        self.conv2 = GCNConv(hidden_channels, hidden_channels*2)
        self.conv3 = GCNConv(hidden_channels*2, hidden_channels)
        self.fc_defect = torch.nn.Linear(hidden_channels, 1)  # 回帰タスク用に出力1次元

    def forward(self, data):
        x, edge_index = data.x.to(device), data.edge_index.to(device)
        x = F.relu(self.conv1(x, edge_index))
        x = F.relu(self.conv2(x, edge_index))
        x = F.relu(self.conv3(x, edge_index))
        return self.fc_defect(x)

# ハイパーパラメータ設定
hidden_channels = 32
learning_rate = 0.0001
batch_size = 32
epochs = 1000
weight_decay = 1e-4

# モデル、損失関数、オプティマイザの定義
model = GCNModel(hidden_channels=hidden_channels).to(device)
optimizer = torch.optim.Adam(model.parameters(), lr=learning_rate, weight_decay=weight_decay)
loss_fn = torch.nn.MSELoss()

# データローダーの作成（ここでデータがすでに定義されていると仮定）
train_loader = DataLoader(train_data_list, batch_size=batch_size, shuffle=True)
val_loader = DataLoader(val_data_list, batch_size=batch_size, shuffle=False)
test_loader = DataLoader(test_data_list, batch_size=batch_size, shuffle=False)

# 学習と検証
train_losses = []
val_losses = []
test_losses = []
start_time = time.time()

for epoch in range(1, epochs + 1):
    model.train()
    total_loss = 0
    for batch in train_loader:
        optimizer.zero_grad()
        out = model(batch).to(device)
        loss = loss_fn(out, batch.y.view(-1, 1).float().to(device))  # yをfloat型に変換してGPUに移動
        loss.backward()
        optimizer.step()
        total_loss += loss.item()
    
    train_losses.append(total_loss / len(train_loader))

    # 検証
    model.eval()
    val_loss = 0
    with torch.no_grad():
        for batch in val_loader:
            out = model(batch).to(device)
            loss = loss_fn(out, batch.y.view(-1, 1).float().to(device))  # yをfloat型に変換してGPUに移動
            val_loss += loss.item()
    
    val_losses.append(val_loss / len(val_loader))

    # 10エポックごとに結果を出力
    if epoch % 10 == 0:
        elapsed_time = time.time() - start_time
        print(f'Epoch {epoch}, Loss: {train_losses[-1]:.4f}, Validation Loss: {val_losses[-1]:.4f}, Elapsed time: {elapsed_time:.2f}s')

# テストデータでの評価
test_loss = 0
test_preds = []
test_targets = []
with torch.no_grad():
    for batch in test_loader:
        out = model(batch).to(device)
        loss = loss_fn(out, batch.y.view(-1, 1).float().to(device))
        test_loss += loss.item()
        test_preds.append(out.cpu().numpy())  # CPUに戻す
        test_targets.append(batch.y.cpu().numpy())

test_loss /= len(test_loader)
test_preds = np.concatenate(test_preds)
test_targets = np.concatenate(test_targets)
rmse = np.sqrt(mean_squared_error(test_targets, test_preds))

print(f'Test Loss: {test_loss:.4f}, RMSE: {rmse:.4f}')

# 損失をプロット
plt.figure(figsize=(12, 6))
plt.plot(range(1, epochs + 1), train_losses, label='Training Loss', color='blue')
plt.plot(range(1, epochs + 1), val_losses, label='Validation Loss', color='orange')
plt.yscale('log')  # 対数スケール
plt.xlabel('Epoch')
plt.ylabel('Loss (Log Scale)')
plt.title('Training and Validation Loss (Log Scale)')
plt.legend()
plt.grid(True, which="both", ls="--")
plt.show()

# 損失をCSVに保存
loss_data = pd.DataFrame({
    'Epoch': range(1, epochs + 1),
    'Training Loss': train_losses,
    'Validation Loss': val_losses
})
loss_data.to_csv(f'loss_data_{test_loss:.4f}.csv', index=False)

test_loss_data = pd.DataFrame({
    'Test Loss': [test_loss],
    'RMSE': [rmse]
})
test_loss_data.to_csv(f'test_loss_{test_loss:.4f}.csv', index=False)

# モデルの保存
torch.save(model.state_dict(), f'gcn_model_{test_loss:.4f}.pth')

# 保存したパラメータを読み込む
model.load_state_dict(torch.load(f'gcn_model_{test_loss:.4f}.pth'))
model.to(device)

print(model)
```
