```python
import os
import time
import json
import numpy as np
import torch
import torch.nn.functional as F
from torch_geometric.nn import GATConv
from torch_geometric.data import Data
from torch_geometric.loader import DataLoader
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import matplotlib.pyplot as plt
import datetime
from collections import defaultdict
from typing import Any, List, Optional, Union
from torch.nn import Module
from torch_geometric.nn.conv import MessagePassing
from torch_geometric.typing import SparseTensor
from torch.utils.tensorboard import SummaryWriter
from torch_geometric.utils import add_self_loops


# # デバイスの設定
# device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
# print(device)
# timestamp = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")

# print(f"Using device: {torch.cuda.get_device_name(0) if torch.cuda.is_available() else 'CPU'}")

# デバイスの設定を明示的に `cuda:0` にする
device = torch.device("cuda:0" if torch.cuda.is_available() else "cpu")
print(device)

timestamp = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")

print(f"Using device: {torch.cuda.get_device_name(0) if torch.cuda.is_available() else 'CPU'}")


# データフォルダ
standardized_data_folder = "/home/nishioka/GNN/Defect_4x4_Normalized1"
label_data_folder = "/home/nishioka/GNN/Defectlabel4x4/DefectLabels_4x4_test1"

# 座標データのロード（最大3654行）
coords = ["x", "y", "z"]
coords_data = {coord: np.load(f"/home/nishioka/GNN/BasicdataforGNN/{coord}_2layer_normalized.npy")[:3654] for coord in coords}

# エッジ情報を読み込む
edges = np.load("/home/nishioka/GNN/BasicdataforGNN/edges_2layer.npy")
edge_index = torch.tensor(edges.T, dtype=torch.long).to(device)

# データとラベルファイルを対応付け
data_files = [f for f in os.listdir(standardized_data_folder) if f.startswith("Normalized1_Defect4x4_ELNOD")]
label_files = [f for f in os.listdir(label_data_folder) if f.startswith("DefectLabel_L")]

# 欠陷なしデータのペア
defect_free_pair = ("/home/nishioka/GNN/BasicdataforGNN/Normalized1_nodefect_ElNOD.npy",
                    "/home/nishioka/GNN/BasicdataforGNN/DefectLabel_nodefect.npy")

# 初期のペアリスト
train_pairs = []
val_pairs = []
test_pairs = []

# 欠陷なしデータを追加
for _ in range(8):
    train_pairs.append(defect_free_pair)
val_pairs.append(defect_free_pair)
test_pairs.append(defect_free_pair)

# データとラベルのペア作成
def extract_layer_block(file_name):
    if file_name.startswith("0"):
        return (0, 0)
    try:
        layer_block_str = file_name.split("_")[-1].replace(".npy", "")
        layer = int(layer_block_str.split("L")[1].split("B")[0])
        block = int(layer_block_str.split("B")[1])
        return (layer, block)
    except (ValueError, IndexError):
        print(f"Invalid file name format: {file_name}")
        return None

data_label_pairs = {}
for data_file in data_files:
    layer_block = extract_layer_block(data_file)
    if layer_block:
        data_label_pairs[layer_block] = {"data": data_file}

for label_file in label_files:
    layer_block = extract_layer_block(label_file)
    if layer_block and layer_block in data_label_pairs:
        data_label_pairs[layer_block]["label"] = label_file

# 有効なペアのみ取得
valid_pairs = [(v["data"], v["label"]) for k, v in data_label_pairs.items() if "label" in v]

# ランダムにサンプリング
num_samples = 1296
random_indices = np.random.choice(len(valid_pairs), num_samples, replace=False)
train_pairs = [valid_pairs[i] for i in random_indices]
remaining_pairs = [valid_pairs[i] for i in range(len(valid_pairs)) if i not in random_indices]
val_pairs, test_pairs = train_test_split(remaining_pairs, test_size=0.5, random_state=42)

# 検証データとテストデータを適切に分けて格納
train_data = [pair[0] for pair in train_pairs]
train_labels = [pair[1] for pair in train_pairs]

val_data = [pair[0] for pair in val_pairs]
val_labels = [pair[1] for pair in val_pairs]

test_data = [pair[0] for pair in test_pairs]
test_labels = [pair[1] for pair in test_pairs]


# データ準備関数
def prepare_data(pairs):
    sampled_data, sampled_labels = [], []
    for data_file, label_file in pairs:
        data_path = os.path.join(standardized_data_folder, data_file)
        label_path = os.path.join(label_data_folder, label_file)
        values = np.load(data_path)[:3654]
        label = np.load(label_path)[:3654]
        node_features = np.vstack((coords_data["x"], coords_data["y"], coords_data["z"], values)).T
        sampled_data.append(node_features)
        sampled_labels.append(label)
    sampled_data = np.stack(sampled_data)
    sampled_labels = np.hstack(sampled_labels)
    x = torch.tensor(sampled_data.reshape(-1, 4), dtype=torch.float).to(device)
    y = torch.tensor(sampled_labels, dtype=torch.float).to(device)
    return x, y

train_x, train_y = prepare_data(train_pairs)
val_x, val_y = prepare_data(val_pairs)
test_x, test_y = prepare_data(test_pairs)

train_data = Data(x=train_x, edge_index=edge_index, y=train_y).to("cuda")
val_data = Data(x=val_x, edge_index=edge_index, y=val_y).to("cuda")
test_data = Data(x=test_x, edge_index=edge_index, y=test_y).to("cuda")

print(train_x.shape, train_y.shape)
print(f"edge_index shape: {edge_index.shape}")
print(f"edge_index: {edge_index}")

# 自己ループを追加
train_data.edge_index, _ = add_self_loops(train_data.edge_index, num_nodes=train_data.x.size(0))
val_data.edge_index, _ = add_self_loops(val_data.edge_index, num_nodes=val_data.x.size(0))
test_data.edge_index, _ = add_self_loops(test_data.edge_index, num_nodes=test_data.x.size(0))

num_nodes = train_data.x.size(0)
print(f"Number of nodes: {num_nodes}")
print(f"Max node index in edge_index: {edge_index.max()}")


print(f"After adding self-loops, train_data.edge_index shape: {train_data.edge_index.shape}")
print(f"After adding self-loops, val_data.edge_index shape: {val_data.edge_index.shape}")
print(f"After adding self-loops, test_data.edge_index shape: {test_data.edge_index.shape}")

# モデル定義
class GATModel(torch.nn.Module):
    def __init__(self, hidden_channels=16):
        super(GATModel, self).__init__()
        self.conv1 = GATConv(4, hidden_channels, heads=2)
        self.conv2 = GATConv(hidden_channels * 2, hidden_channels, heads=1)
        self.conv3 = GATConv(hidden_channels, hidden_channels, heads=1)
        self.fc_defect = torch.nn.Linear(hidden_channels, 1)

    def forward(self, data):
        x, edge_index = data.x, data.edge_index
        x = F.relu(self.conv1(x, edge_index))
        x = F.relu(self.conv2(x, edge_index))
        x = F.relu(self.conv3(x, edge_index))
        x = self.fc_defect(x)
        return torch.sigmoid(x)

model = GATModel(hidden_channels=16).to(device)
if torch.cuda.device_count() > 1:
    print(f"Using {torch.cuda.device_count()} GPUs with Data Parallel")
    model = torch.nn.DataParallel(model, device_ids=[0, 1])
    model.to(device)

model.apply(lambda m: torch.nn.init.xavier_uniform_(m.weight) if isinstance(m, torch.nn.Linear) else None)
optimizer = torch.optim.Adam(model.parameters(), lr=0.001, weight_decay=5e-4)
loss_fn = torch.nn.MSELoss()


# データをサンプルごとに分割
train_data_list = []
val_data_list = []
test_data_list = []

# サンプル単位でデータをDataLoaderに渡す
num_train_samples = train_x.shape[0] // 3654  # サンプル数の計算
num_val_samples = val_x.shape[0] // 3654
num_test_samples = test_x.shape[0] // 3654

for i in range(num_train_samples):
    train_data_list.append(Data(x=train_x[i*3654:(i+1)*3654], edge_index=edge_index, y=train_y[i*3654:(i+1)*3654]))

for i in range(num_val_samples):
    val_data_list.append(Data(x=val_x[i*3654:(i+1)*3654], edge_index=edge_index, y=val_y[i*3654:(i+1)*3654]))

for i in range(num_test_samples):
    test_data_list.append(Data(x=test_x[i*3654:(i+1)*3654], edge_index=edge_index, y=test_y[i*3654:(i+1)*3654]))

# データローダーの作成
train_loader = DataLoader(train_data_list, batch_size=32, shuffle=True)
val_loader = DataLoader(val_data_list, batch_size=32, shuffle=False)
test_loader = DataLoader(test_data_list, batch_size=32, shuffle=False)

print(f"Batch x shape: {batch.x.shape}")
print(f"Batch edge_index shape: {batch.edge_index.shape}")
print(f"Batch y shape: {batch.y.shape}")
print(f"Batch edge_index: {batch.edge_index}")

# バッチの形状を確認する
for batch in train_loader:
    print(f"Batch x shape batch: {batch.x.shape}")
    print(f"Batch edge_index shape: {batch.edge_index.shape}")
    print(f"Batch y shape: {batch.y.shape}")
    print(f"Batch edge_index: {batch.edge_index}")
    break  # 最初のバッチだけを表示するためにbreakを使います


for epoch in range(1, 2001):
    model.train()
    train_loss = 0
    for batch in train_loader:
        # データをデバイスに移動
        batch.x = batch.x.to(device)
        batch.edge_index = batch.edge_index.to(device)
        batch.y = batch.y.to(device)

        # エッジインデックスの範囲を確認
        num_nodes = batch.x.size(0)
        max_node_index = batch.edge_index.max()
        print(f"Number of nodes: {num_nodes}, Max node index in edge_index: {max_node_index}")

        if max_node_index >= num_nodes:
            print("Warning: edge_index contains indices that are out of bounds!")

        # 自己ループを追加
        batch.edge_index, _ = add_self_loops(batch.edge_index, num_nodes=num_nodes)
        batch.edge_index, _ = add_self_loops(batch.edge_index, num_nodes=batch.x.size(0))

        try:
            out = model(batch)
        except IndexError as e:
            print(f"IndexError: {e}")
            print(f"Batch x shape: {batch.x.shape}")
            print(f"Batch edge_index shape: {batch.edge_index.shape}")
            print(f"Max node index in edge_index: {batch.edge_index.max()}")


        # 損失の計算
        loss = loss_fn(out, batch.y.view(-1, 1))
        loss.backward()
        optimizer.step()
        train_loss += loss.item()

    
    print(f'Epoch {epoch}, Train Loss: {train_loss/len(train_loader)}')

    # 検証データでも同様に形状を確認（毎エポックでなくてもよい）
    model.eval()
    with torch.no_grad():
        for batch in val_loader:
            batch.x = batch.x.to(device)
            batch.y = batch.y.to(device)
            batch.edge_index = batch.edge_index.to(device)
            print(f"\n[Validation Epoch {epoch}]")
            print(f"Val Batch x shape: {batch.x.shape}")
            print(f"Val Batch edge_index shape: {batch.edge_index.shape}")
            print(f"Val Batch y shape: {batch.y.shape}")

            out = model(batch)
            
            print(f"Val Model output shape: {out.shape}")
            print(f"Val Model output: {out[:5]}")

            # テストデータでの評価
            test_loss = 0
            test_preds = []
            test_targets = []
            with torch.no_grad():
                for batch in test_loader:
                    batch.x = batch.x.to(device)
                    batch.y = batch.y.to(device)
                    batch.edge_index = batch.edge_index.to(device)
                    
                    out = model(batch.x, batch.edge_index)
                    
                    loss = loss_fn(out, batch.y.view(-1, 1))
                    test_loss += loss.item()
                    test_preds.append(out.cpu().numpy())
                    test_targets.append(batch.y.cpu().numpy())
            
            test_loss /= len(test_loader)
            test_preds = np.concatenate(test_preds)
            test_targets = np.concatenate(test_targets)
            
            # RMSE (Root Mean Squared Error)
            rmse = np.sqrt(mean_squared_error(test_targets, test_preds))
            
            # MAE (Mean Absolute Error)
            mae = mean_absolute_error(test_targets, test_preds)
            
            # R2 Score
            r2 = r2_score(test_targets, test_preds)
            
            print(f'Test Loss: {test_loss:.4f}, RMSE: {rmse:.4f}, MAE: {mae:.4f}, R2 Score: {r2:.4f}')

print(summary(model, x, edge_index))

# カスタムの summary 関数を定義
def summary(model: torch.nn.Module, *args, max_depth: int = 3,
            leaf_module: Optional[Union[Module, List[Module]]] = 'MessagePassing', **kwargs) -> str:
    if leaf_module == 'MessagePassing':
        leaf_module = MessagePassing

    def register_hook(info):
        def hook(module, inputs, output):
            info['input_shape'].append(get_shape(inputs))
            info['output_shape'].append(get_shape(output))
        return hook

    hooks = {}
    depth = 0
    stack = [(model.__class__.__name__, model, depth)]
    info_list = []
    input_shape = defaultdict(list)
    output_shape = defaultdict(list)
    while stack:
        name, module, depth = stack.pop()
        module_id = id(module)
        if name.startswith('(_'):
            continue
        if module_id in hooks:
            hooks[module_id].remove()

        info = {
            'name': name,
            'input_shape': input_shape[module_id],
            'output_shape': output_shape[module_id],
            'depth': depth,
            '#param': f'{sum(p.numel() for p in module.parameters()):,}' if len(list(module.parameters())) > 0 else '--'
        }

        if depth >= max_depth or isinstance(module, leaf_module):
            continue

        module_items = reversed(module._modules.items())
        stack += [(f"({name}){mod.__class__.__name__}", mod, depth + 1) for name, mod in module_items if mod is not None]

    # モデルの forward pass のフックを設定
    training = model.training
    model.eval()
    with torch.no_grad():
        model(*args, **kwargs)
    model.train(training)

    for h in hooks.values():
        h.remove()

    info_list = postprocess(info_list)
    return make_table(info_list, max_depth=max_depth)

# サマリー表示のためのヘルパー関数を定義
def get_shape(inputs: Any) -> str:
    if not isinstance(inputs, (tuple, list)):
        inputs = (inputs,)
    out = []
    for x in inputs:
        if isinstance(x, SparseTensor):
            out.append(str(list(x.sizes())))
        elif hasattr(x, 'size'):
            out.append(str(list(x.size())))
    return ', '.join(out)

def postprocess(info_list: List[dict]) -> List[dict]:
    for idx, info in enumerate(info_list):
        depth = info['depth']
        prefix = '├─' if depth == 1 else f"{'│    '*(depth-1)}└─"
        info['name'] = prefix + info['name']
        if info['input_shape']:
            info['input_shape'] = info['input_shape'][0]
            info['output_shape'] = info['output_shape'][0]
        else:
            info['input_shape'] = '--'
            info['output_shape'] = '--'
    return info_list

def make_table(info_list: List[dict], max_depth: int) -> str:
    from tabulate import tabulate
    content = [['Layer', 'Input Shape', 'Output Shape', '#Param']]
    for info in info_list:
        content.append([info['name'], info['input_shape'], info['output_shape'], info['#param']])
    return tabulate(content, headers='firstrow', tablefmt='psql')



```
