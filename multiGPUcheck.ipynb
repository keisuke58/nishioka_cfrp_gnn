```python
import os
import torch

# Set default hyperparameters directly if running in an interactive environment
learning_rate = 0.00001
batch_size = 64
epochs = 100
gpu_id = -1  # Automatically assign GPU if -1

# Function to check how many GPUs are available
def get_available_gpus():
    available_gpus = torch.cuda.device_count()
    print(f"Number of available GPUs: {available_gpus}")
    return available_gpus

# Automatically assign GPU if none is specified
if gpu_id == -1:
    available_gpus = get_available_gpus()
    if available_gpus > 0:
        # Dynamically assign the GPU based on available GPU count
        gpu_id = 0  # For Jupyter, we select the first GPU
        os.environ['CUDA_VISIBLE_DEVICES'] = str(gpu_id)
        print(f"Assigned GPU {gpu_id}")
        device = torch.device("cuda")
    else:
        print("No GPU available, using CPU")
        device = torch.device("cpu")
else:
    os.environ['CUDA_VISIBLE_DEVICES'] = str(gpu_id)
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
    print(f"Manually assigned GPU {gpu_id}")

# Print device information
print(f"Using device: {device}")
print(f"Running with learning rate: {learning_rate}, batch size: {batch_size}, epochs: {epochs}")

# Rest of the model code, using the assigned device...
```
