```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import confusion_matrix
from matplotlib.colors import ListedColormap
from matplotlib import ticker
from matplotlib.backends.backend_pdf import PdfPages

# Generate random labels for testing
np.random.seed(42)
num_classes = 19
all_labels = np.random.randint(0, num_classes, 1000)
all_preds = np.random.randint(0, num_classes, 1000)

# Compute confusion matrix
cm = confusion_matrix(all_labels, all_preds)

# Plotting the confusion matrix with enhanced visualization
fig, ax = plt.subplots(figsize=(12, 10))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", cbar=False, linewidths=.5,
            square=True, annot_kws={"size": 10}, ax=ax)

# Set tick labels and axis labels for improved readability
tick_labels = [f"Class {i}" for i in range(num_classes)]
ax.set_xticks(np.arange(num_classes) + 0.5)
ax.set_xticklabels(tick_labels, rotation=45, ha="right", fontsize=10, weight="bold")
ax.set_yticks(np.arange(num_classes) + 0.5)
ax.set_yticklabels(tick_labels, rotation=0, fontsize=10, weight="bold")

# Set axis titles with improved visibility
ax.set_xlabel("Predicted", fontsize=12, weight="bold", labelpad=10)
ax.set_ylabel("Actual", fontsize=12, weight="bold", labelpad=10)
plt.title("Confusion Matrix", fontsize=15, weight="bold")

# Set color intensity in grid for a cleaner look
for _, spine in ax.spines.items():
    spine.set_visible(True)
    spine.set_linewidth(1.5)

plt.tight_layout()
plt.show()
```
