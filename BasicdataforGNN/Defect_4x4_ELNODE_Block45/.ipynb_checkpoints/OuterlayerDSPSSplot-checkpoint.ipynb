```python
import re
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec

# Path to the .rpt file
file_path = "ELCENT_defect_4x4_Layer16_Block45.rpt"

# Function to extract data for a specific region
def extract_data(start_element, end_element, region_name):
    element_labels = []
    sum_principal_stresses = []

    # Regular expression pattern to extract element label and principal stresses
    pattern = re.compile(r'^\s*(\d+)\s+([-\d.E+]+)\s+([-\d.E+]+)\s+([-\d.E+]+)')

    # Flag to control when to start reading
    start_reading = False

    with open(file_path, 'r') as file:
        for line in file:
            if f"Field Output reported at element centroid for region: {region_name}" in line:
                start_reading = True
                continue
            
            if start_reading and "-----------------------------------------------------------------" in line:
                start_reading = True
                continue
            
            if start_reading:
                match = pattern.match(line)
                if match:
                    element_label = int(match.group(1))
                    max_principal = float(match.group(2).replace('E', 'e'))
                    mid_principal = float(match.group(3).replace('E', 'e'))
                    min_principal = float(match.group(4).replace('E', 'e'))
                    
                    # Calculate the sum of the principal stresses
                    sum_stresses = max_principal + mid_principal + min_principal
                    
                    # Store the values
                    element_labels.append(element_label)
                    sum_principal_stresses.append(sum_stresses)

    # Convert to numpy arrays for easier manipulation
    element_labels = np.array(element_labels)
    sum_principal_stresses = np.array(sum_principal_stresses)

    # Define the grid size (same for both regions)
    nrows, ncols = 28, 62

    # Generate the expected element labels grid based on the starting and ending element numbers
    expected_element_labels = np.arange(start_element, end_element + 1).reshape(nrows, ncols, order='F')

    # Create an array to store the principal stresses with NaN as default values
    reshaped_stress_data = np.full(expected_element_labels.shape, np.nan)

    # Place actual stress values into the correct positions, matching element numbers
    for i, elem in enumerate(element_labels):
        index = np.where(expected_element_labels == elem)
        reshaped_stress_data[index] = sum_principal_stresses[i]
    
    return reshaped_stress_data

# Define start_element and end_element for both regions
def get_element_range(region_name):
    if region_name == "INTERSTAGE_1/8-1.Region_21":
        return 34721, 36456
    elif region_name == "INTERSTAGE_1/8-1.Region_1":
        return 1, 1736
    else:
        raise ValueError(f"Unknown region: {region_name}")

# Calculate horizontal and vertical profiles
def calculate_profiles(reshaped_stress_data):
    # Horizontal Profile (rows 11 to 14)
    rows_to_average = reshaped_stress_data[10:14, :]
    horizontal_profile = np.nanmean(rows_to_average, axis=0)

    # Vertical Profile (columns 29, 30, 31, 32)
    columns_to_average = reshaped_stress_data[:, 28:32]
    vertical_profile = np.nanmean(columns_to_average, axis=1)
    
    return horizontal_profile, vertical_profile

# Extract data for both regions
start_element, end_element = get_element_range("INTERSTAGE_1/8-1.Region_21")
reshaped_stress_data_region_21 = extract_data(start_element, end_element, "INTERSTAGE_1/8-1.Region_21")

start_element, end_element = get_element_range("INTERSTAGE_1/8-1.Region_1")
reshaped_stress_data_region_1 = extract_data(start_element, end_element, "INTERSTAGE_1/8-1.Region_1")

# Calculate horizontal and vertical profiles for both regions
horizontal_profile_region_21, vertical_profile_region_21 = calculate_profiles(reshaped_stress_data_region_21)
horizontal_profile_region_1, vertical_profile_region_1 = calculate_profiles(reshaped_stress_data_region_1)

# Create a new figure and define the GridSpec layout
fig = plt.figure(figsize=(14.4, 6.8))  # Adjust the layout size
gs = gridspec.GridSpec(2, 4, width_ratios=[1, 5, 1, 5], height_ratios=[3, 5])

# Title for Region 21
ax_title_21 = plt.subplot(gs[0, 1])
ax_title_21.set_title('DSPSS Region_21', fontsize=14)
ax_title_21.axis('off')

# Title for Region 1
ax_title_1 = plt.subplot(gs[0, 3])
ax_title_1.set_title('DSPSS Region_1', fontsize=14)
ax_title_1.axis('off')

# Heatmap for Region 21
ax_heatmap_21 = plt.subplot(gs[1, 1])
im_21 = ax_heatmap_21.imshow(reshaped_stress_data_region_21, cmap='viridis', aspect='auto')
fig.colorbar(im_21, ax=ax_heatmap_21, location='left', label='DSPSS [MPa]')

# Heatmap for Region 1
ax_heatmap_1 = plt.subplot(gs[1, 3])
im_1 = ax_heatmap_1.imshow(reshaped_stress_data_region_1, cmap='viridis', aspect='auto')
fig.colorbar(im_1, ax=ax_heatmap_1, location='left', label='DSPSS [MPa]')

# Horizontal profile for Region 21
ax_horizontal_21 = plt.subplot(gs[0, 1], sharex=ax_heatmap_21)
ax_horizontal_21.plot(horizontal_profile_region_21)
ax_horizontal_21.set_title('Horizontal Profile Region_21')
ax_horizontal_21.set_ylabel('Mean DSPSS')
ax_horizontal_21.set_xlim([0, len(horizontal_profile_region_21)])
ax_horizontal_21.tick_params(labelbottom=False)

# Horizontal profile for Region 1
ax_horizontal_1 = plt.subplot(gs[0, 3], sharex=ax_heatmap_1)
ax_horizontal_1.plot(horizontal_profile_region_1)
ax_horizontal_1.set_title('Horizontal Profile Region_1')
ax_horizontal_1.set_ylabel('Mean DSPSS')
ax_horizontal_1.set_xlim([0, len(horizontal_profile_region_1)])
ax_horizontal_1.tick_params(labelbottom=False)

# Vertical profile for Region 21
ax_vertical_21 = plt.subplot(gs[1, 0], sharey=ax_heatmap_21)
ax_vertical_21.plot(vertical_profile_region_21, range(len(vertical_profile_region_21)))
ax_vertical_21.set_title('Vertical Profile Region_21', loc='left', rotation=270)
ax_vertical_21.set_xlabel('Mean DSPSS')

# Vertical profile for Region 1
ax_vertical_1 = plt.subplot(gs[1, 2], sharey=ax_heatmap_1)
ax_vertical_1.plot(vertical_profile_region_1, range(len(vertical_profile_region_1)))
ax_vertical_1.set_title('Vertical Profile Region_1', loc='left', rotation=270)
ax_vertical_1.set_xlabel('Mean DSPSS')

# Show the plot
plt.subplots_adjust(left=0.1, right=0.9, top=0.9, bottom=0.1, wspace=0.4, hspace=0.4)
plt.tight_layout()
plt.show()
```

```python
import re
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec

# Path to the .rpt file
file_path = "ELCENT_defect_4x4_Layer16_Block45.rpt"


# Define the grid size
nrows, ncols = 28, 62

# Generate the expected element labels grid for Region_21 (normal order)
expected_element_labels_region_21 = np.arange(34721, 36456 + 1).reshape(nrows, ncols, order='F')

# For Region_1, reverse the vertical order of the elements
expected_element_labels_region_1 = np.flipud(np.arange(1, 1736 + 1).reshape(nrows, ncols, order='F'))

# Modify the extract_data function to handle both regions
def extract_data(start_element, end_element, region_name):
    element_labels = []
    sum_principal_stresses = []

    # Regular expression pattern to extract element label and principal stresses
    pattern = re.compile(r'^\s*(\d+)\s+([-\d.E+]+)\s+([-\d.E+]+)\s+([-\d.E+]+)')

    # Flag to control when to start reading
    start_reading = False

    with open(file_path, 'r') as file:
        for line in file:
            if f"Field Output reported at element centroid for region: {region_name}" in line:
                start_reading = True
                continue
            
            if start_reading and "-----------------------------------------------------------------" in line:
                start_reading = True
                continue
            
            if start_reading:
                match = pattern.match(line)
                if match:
                    element_label = int(match.group(1))
                    max_principal = float(match.group(2).replace('E', 'e'))
                    mid_principal = float(match.group(3).replace('E', 'e'))
                    min_principal = float(match.group(4).replace('E', 'e'))
                    
                    # Calculate the sum of the principal stresses
                    sum_stresses = max_principal + mid_principal + min_principal
                    
                    # Store the values
                    element_labels.append(element_label)
                    sum_principal_stresses.append(sum_stresses)

    # Convert to numpy arrays for easier manipulation
    element_labels = np.array(element_labels)
    sum_principal_stresses = np.array(sum_principal_stresses)

    # Use the appropriate element label grid for each region
    if region_name == "INTERSTAGE_1/8-1.Region_21":
        expected_element_labels = expected_element_labels_region_21
    elif region_name == "INTERSTAGE_1/8-1.Region_1":
        expected_element_labels = expected_element_labels_region_1
    else:
        raise ValueError(f"Unknown region: {region_name}")

    # Create an array to store the principal stresses with NaN as default values
    reshaped_stress_data = np.full(expected_element_labels.shape, np.nan)

    # Place actual stress values into the correct positions, matching element numbers
    for i, elem in enumerate(element_labels):
        index = np.where(expected_element_labels == elem)
        reshaped_stress_data[index] = sum_principal_stresses[i]

    return reshaped_stress_data
     

# Define start_element and end_element for both regions
def get_element_range(region_name):
    if region_name == "INTERSTAGE_1/8-1.Region_21":
        return 34721, 36456
    elif region_name == "INTERSTAGE_1/8-1.Region_1":
        return 1, 1736
    else:
        raise ValueError(f"Unknown region: {region_name}")

# Calculate horizontal and vertical profiles
def calculate_profiles(reshaped_stress_data):
    # Horizontal Profile (rows 11 to 14)
    rows_to_average = reshaped_stress_data[10:14, :]
    horizontal_profile = np.nanmean(rows_to_average, axis=0)

    # Vertical Profile (columns 29, 30, 31, 32)
    columns_to_average = reshaped_stress_data[:, 28:32]
    vertical_profile = np.nanmean(columns_to_average, axis=1)
    
    return horizontal_profile, vertical_profile

# Extract data for both regions
start_element, end_element = get_element_range("INTERSTAGE_1/8-1.Region_21")
reshaped_stress_data_region_21 = extract_data(start_element, end_element, "INTERSTAGE_1/8-1.Region_21")

start_element, end_element = get_element_range("INTERSTAGE_1/8-1.Region_1")
reshaped_stress_data_region_1 = extract_data(start_element, end_element, "INTERSTAGE_1/8-1.Region_1")

# Calculate horizontal and vertical profiles for both regions
horizontal_profile_region_21, vertical_profile_region_21 = calculate_profiles(reshaped_stress_data_region_21)
horizontal_profile_region_1, vertical_profile_region_1 = calculate_profiles(reshaped_stress_data_region_1)

# Create a new figure and define the GridSpec layout
fig = plt.figure(figsize=(14.4, 6.8))  # Adjust the layout size
gs = gridspec.GridSpec(2, 4, width_ratios=[1, 5, 1, 5], height_ratios=[3, 5])

# Title for Region 21
ax_title_21 = plt.subplot(gs[0, 1])
ax_title_21.set_title('DSPSS Region_21', fontsize=14)
ax_title_21.axis('off')

# Title for Region 1
ax_title_1 = plt.subplot(gs[0, 3])
ax_title_1.set_title('DSPSS Region_1', fontsize=14)
ax_title_1.axis('off')

# Heatmap for Region 21
ax_heatmap_21 = plt.subplot(gs[1, 1])
im_21 = ax_heatmap_21.imshow(reshaped_stress_data_region_21, cmap='viridis', aspect='auto')
fig.colorbar(im_21, ax=ax_heatmap_21, location='left', label='DSPSS [MPa]')

# Heatmap for Region 1
ax_heatmap_1 = plt.subplot(gs[1, 3])
im_1 = ax_heatmap_1.imshow(reshaped_stress_data_region_1, cmap='viridis', aspect='auto')
fig.colorbar(im_1, ax=ax_heatmap_1, location='left', label='DSPSS [MPa]')

# Horizontal profile for Region 21
ax_horizontal_21 = plt.subplot(gs[0, 1], sharex=ax_heatmap_21)
ax_horizontal_21.plot(horizontal_profile_region_21)
ax_horizontal_21.set_title('Horizontal Profile Region_21')
ax_horizontal_21.set_ylabel('Mean DSPSS')
ax_horizontal_21.set_xlim([0, len(horizontal_profile_region_21)])
ax_horizontal_21.tick_params(labelbottom=False)

# Horizontal profile for Region 1
ax_horizontal_1 = plt.subplot(gs[0, 3], sharex=ax_heatmap_1)
ax_horizontal_1.plot(horizontal_profile_region_1)
ax_horizontal_1.set_title('Horizontal Profile Region_1')
ax_horizontal_1.set_ylabel('Mean DSPSS')
ax_horizontal_1.set_xlim([0, len(horizontal_profile_region_1)])
ax_horizontal_1.tick_params(labelbottom=False)

# Vertical profile for Region 21
ax_vertical_21 = plt.subplot(gs[1, 0], sharey=ax_heatmap_21)
ax_vertical_21.plot(vertical_profile_region_21, range(len(vertical_profile_region_21)))
ax_vertical_21.set_title('Vertical Profile Region_21', loc='left', rotation=270)
ax_vertical_21.set_xlabel('Mean DSPSS')

# Vertical profile for Region 1
ax_vertical_1 = plt.subplot(gs[1, 2], sharey=ax_heatmap_1)
ax_vertical_1.plot(vertical_profile_region_1, range(len(vertical_profile_region_1)))
ax_vertical_1.set_title('Vertical Profile Region_1', loc='left', rotation=270)
ax_vertical_1.set_xlabel('Mean DSPSS')

# Show the plot
plt.subplots_adjust(left=0.1, right=0.9, top=0.9, bottom=0.1, wspace=0.4, hspace=0.4)
plt.tight_layout()
plt.show()
```

```python
import re
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec

# Path to the .rpt file
file_path = "ELCENT_defect_4x4_Layer16_Block45.rpt"

# Function to extract data for a specific region
def extract_data(start_element, end_element, region_name):
    element_labels = []
    sum_principal_stresses = []

    # Regular expression pattern to extract element label and principal stresses
    pattern = re.compile(r'^\s*(\d+)\s+([-\d.E+]+)\s+([-\d.E+]+)\s+([-\d.E+]+)')

    # Flag to control when to start reading
    start_reading = False

    with open(file_path, 'r') as file:
        for line in file:
            if f"Field Output reported at element centroid for region: {region_name}" in line:
                start_reading = True
                continue
            
            if start_reading and "-----------------------------------------------------------------" in line:
                start_reading = True
                continue
            
            if start_reading:
                match = pattern.match(line)
                if match:
                    element_label = int(match.group(1))
                    max_principal = float(match.group(2).replace('E', 'e'))
                    mid_principal = float(match.group(3).replace('E', 'e'))
                    min_principal = float(match.group(4).replace('E', 'e'))
                    
                    # Calculate the sum of the principal stresses
                    sum_stresses = max_principal + mid_principal + min_principal
                    
                    # Store the values
                    element_labels.append(element_label)
                    sum_principal_stresses.append(sum_stresses)

    # Convert to numpy arrays for easier manipulation
    element_labels = np.array(element_labels)
    sum_principal_stresses = np.array(sum_principal_stresses)

    # Define the grid size (same for both regions)
    nrows, ncols = 28, 62

    # Generate the expected element labels grid based on the starting and ending element numbers
    if region_name == "INTERSTAGE_1/8-1.Region_1":
        expected_element_labels = np.flipud(np.arange(start_element, end_element + 1).reshape(nrows, ncols, order='F'))
    else:
        expected_element_labels = np.arange(start_element, end_element + 1).reshape(nrows, ncols, order='F')

    # Create an array to store the principal stresses with NaN as default values
    reshaped_stress_data = np.full(expected_element_labels.shape, np.nan)

    # Place actual stress values into the correct positions, matching element numbers
    for i, elem in enumerate(element_labels):
        index = np.where(expected_element_labels == elem)
        reshaped_stress_data[index] = sum_principal_stresses[i]

    return reshaped_stress_data

# Define start_element and end_element for both regions
def get_element_range(region_name):
    if region_name == "INTERSTAGE_1/8-1.Region_21":
        return 34721, 36456
    elif region_name == "INTERSTAGE_1/8-1.Region_1":
        return 1, 1736
    else:
        raise ValueError(f"Unknown region: {region_name}")

# Calculate horizontal and vertical profiles
def calculate_profiles(reshaped_stress_data):
    # Horizontal Profile (rows 11 to 14)
    rows_to_average = reshaped_stress_data[10:14, :]
    horizontal_profile = np.nanmean(rows_to_average, axis=0)

    # Vertical Profile (columns 29, 30, 31, 32)
    columns_to_average = reshaped_stress_data[:, 28:32]
    vertical_profile = np.nanmean(columns_to_average, axis=1)
    
    return horizontal_profile, vertical_profile

# Set the common DSPSS range for all plots
vmin, vmax = 1600, 2100  # You can adjust these values as needed

# Extract data for both regions
start_element, end_element = get_element_range("INTERSTAGE_1/8-1.Region_21")
reshaped_stress_data_region_21 = extract_data(start_element, end_element, "INTERSTAGE_1/8-1.Region_21")

start_element, end_element = get_element_range("INTERSTAGE_1/8-1.Region_1")
reshaped_stress_data_region_1 = extract_data(start_element, end_element, "INTERSTAGE_1/8-1.Region_1")

# Calculate horizontal and vertical profiles for both regions
horizontal_profile_region_21, vertical_profile_region_21 = calculate_profiles(reshaped_stress_data_region_21)
horizontal_profile_region_1, vertical_profile_region_1 = calculate_profiles(reshaped_stress_data_region_1)

# Create a new figure and define the GridSpec layout
fig = plt.figure(figsize=(14.4, 6.8))  # Adjust the layout size
gs = gridspec.GridSpec(2, 4, width_ratios=[1, 5, 1, 5], height_ratios=[3, 5])

# Title for Region 21
ax_title_21 = plt.subplot(gs[0, 1])
ax_title_21.set_title('DSPSS Region_21', fontsize=14)
ax_title_21.axis('off')

# Title for Region 1
ax_title_1 = plt.subplot(gs[0, 3])
ax_title_1.set_title('DSPSS Region_1', fontsize=14)
ax_title_1.axis('off')

# Heatmap for Region 21
ax_heatmap_21 = plt.subplot(gs[1, 1])
im_21 = ax_heatmap_21.imshow(reshaped_stress_data_region_21, cmap='viridis', aspect='auto', vmin=vmin, vmax=vmax)
fig.colorbar(im_21, ax=ax_heatmap_21, location='left', label='DSPSS [MPa]')

# Heatmap for Region 1
ax_heatmap_1 = plt.subplot(gs[1, 3])
im_1 = ax_heatmap_1.imshow(reshaped_stress_data_region_1, cmap='viridis', aspect='auto', vmin=vmin, vmax=vmax)
fig.colorbar(im_1, ax=ax_heatmap_1, location='left', label='DSPSS [MPa]')

# Horizontal profile for Region 21
ax_horizontal_21 = plt.subplot(gs[0, 1], sharex=ax_heatmap_21)
ax_horizontal_21.plot(horizontal_profile_region_21)
ax_horizontal_21.set_title('Horizontal Profile Region_21')
ax_horizontal_21.set_ylabel('Mean DSPSS')
ax_horizontal_21.set_xlim([0, len(horizontal_profile_region_21)])  # Set x-axis limits to match the heatmap
ax_horizontal_21.set_ylim([vmin, vmax])  # Set y-axis limits for the horizontal profile
ax_horizontal_21.tick_params(labelbottom=False)

# Horizontal profile for Region 1
ax_horizontal_1 = plt.subplot(gs[0, 3], sharex=ax_heatmap_1)
ax_horizontal_1.plot(horizontal_profile_region_1)
ax_horizontal_1.set_title('Horizontal Profile Region_1')
ax_horizontal_1.set_ylabel('Mean DSPSS')
ax_horizontal_1.set_xlim([0, len(horizontal_profile_region_1)])
ax_horizontal_1.set_ylim([vmin, vmax])  # Set y-axis limits for the horizontal profile
ax_horizontal_1.tick_params(labelbottom=False)

# Vertical profile for Region 21
ax_vertical_21 = plt.subplot(gs[1, 0], sharey=ax_heatmap_21)
ax_vertical_21.plot(vertical_profile_region_21, range(len(vertical_profile_region_21)))
# ax_vertical_21.set_title('Vertical Profile Region_21', loc='left', rotation=270)
ax_vertical_21.set_xlabel('Mean DSPSS')
ax_vertical_21.set_xlim([vmin, vmax])  # Set y-axis limits for the vertical profile

# Vertical profile for Region 1
ax_vertical_1 = plt.subplot(gs[1, 2], sharey=ax_heatmap_1)
ax_vertical_1.plot(vertical_profile_region_1, range(len(vertical_profile_region_1)))
# ax_vertical_1.set_title('Vertical Profile Region_1', loc='left', rotation=270)
ax_vertical_1.set_xlabel('Mean DSPSS')
ax_vertical_1.set_xlim([vmin, vmax])  # Set y-axis limits for the vertical profile

# Show the plot
plt.subplots_adjust(left=0.1, right=0.9, top=0.9, bottom=0.1, wspace=0.4, hspace=0.4)
plt.tight_layout()
plt.show()
```
