```python
import re
import numpy as np
import matplotlib.pyplot as plt

# Function to extract data for a specific region
def extract_data(file_path, start_element, end_element, region_name):
    element_labels = []
    sum_principal_stresses = []

    # Regular expression pattern to extract element label and principal stresses
    pattern = re.compile(r'^\s*(\d+)\s+([-\d.E+]+)\s+([-\d.E+]+)\s+([-\d.E+]+)')

    # Flag to control when to start reading
    start_reading = False

    with open(file_path, 'r') as file:
        for line in file:
            if f"Field Output reported at element centroid for region: {region_name}" in line:
                start_reading = True
                continue
            
            if start_reading and "-----------------------------------------------------------------" in line:
                start_reading = True
                continue
            
            if start_reading:
                match = pattern.match(line)
                if match:
                    element_label = int(match.group(1))
                    max_principal = float(match.group(2).replace('E', 'e'))
                    mid_principal = float(match.group(3).replace('E', 'e'))
                    min_principal = float(match.group(4).replace('E', 'e'))
                    
                    # Calculate the sum of the principal stresses
                    sum_stresses = max_principal + mid_principal + min_principal
                    
                    # Store the values
                    element_labels.append(element_label)
                    sum_principal_stresses.append(sum_stresses)

    # Convert to numpy arrays for easier manipulation
    element_labels = np.array(element_labels)
    sum_principal_stresses = np.array(sum_principal_stresses)

    # Define the grid size (same for both regions)
    nrows, ncols = 28, 62

    # Generate the expected element labels grid based on the starting and ending element numbers
    expected_element_labels = np.arange(start_element, end_element + 1).reshape(nrows, ncols, order='F')

    # Create an array to store the principal stresses with NaN as default values
    reshaped_stress_data = np.full(expected_element_labels.shape, np.nan)

    # Place actual stress values into the correct positions, matching element numbers
    for i, elem in enumerate(element_labels):
        index = np.where(expected_element_labels == elem)
        reshaped_stress_data[index] = sum_principal_stresses[i]

    return reshaped_stress_data

# Calculate horizontal and vertical profiles
def calculate_profiles(reshaped_stress_data):
    # Horizontal Profile (rows 11 to 14)
    rows_to_average = reshaped_stress_data[10:14, :]
    horizontal_profile = np.nanmean(rows_to_average, axis=0)

    # Vertical Profile (columns 29, 30, 31, 32)
    columns_to_average = reshaped_stress_data[:, 28:32]
    vertical_profile = np.nanmean(columns_to_average, axis=1)
    
    return horizontal_profile, vertical_profile

# Lists to store profiles for each layer
horizontal_profiles = []
vertical_profiles = []

# Loop through each layer and extract profiles
for i in range(11, 20):
    file_path = f"ELCENT_defect_4x4_Layer{i}_Block45.rpt"
    start_element, end_element = 34721, 36456  # Example values, adjust as needed
    reshaped_stress_data = extract_data(file_path, start_element, end_element, f"INTERSTAGE_1/8-1.Region_21")
    
    # Calculate profiles
    horizontal_profile, vertical_profile = calculate_profiles(reshaped_stress_data)
    
    # Store profiles
    horizontal_profiles.append(horizontal_profile)
    vertical_profiles.append(vertical_profile)

# Plot horizontal profiles
plt.figure(figsize=(10, 6))
for i, profile in enumerate(horizontal_profiles, start=11):
    plt.plot(profile, label=f'Layer {i}')
plt.title('Horizontal Profiles')
plt.xlabel('Width (Columns)')
plt.ylabel('Mean DSPSS')
plt.legend()
plt.show()

# Plot vertical profiles
plt.figure(figsize=(10, 6))
for i, profile in enumerate(vertical_profiles, start=11):
    plt.plot(profile, label=f'Layer {i}')
plt.title('Vertical Profiles')
plt.xlabel('Height (Rows)')
plt.ylabel('Mean DSPSS')
plt.legend()
plt.show()
```

```python
import re
import numpy as np
import matplotlib.pyplot as plt

# Function to extract data for a specific region
def extract_data(file_path, start_element, end_element, region_name):
    element_labels = []
    sum_principal_stresses = []

    # Regular expression pattern to extract element label and principal stresses
    pattern = re.compile(r'^\s*(\d+)\s+([-\d.E+]+)\s+([-\d.E+]+)\s+([-\d.E+]+)')

    # Flag to control when to start reading
    start_reading = False

    with open(file_path, 'r') as file:
        for line in file:
            if f"Field Output reported at element centroid for region: {region_name}" in line:
                start_reading = True
                continue
            
            if start_reading and "-----------------------------------------------------------------" in line:
                start_reading = True
                continue
            
            if start_reading:
                match = pattern.match(line)
                if match:
                    element_label = int(match.group(1))
                    max_principal = float(match.group(2).replace('E', 'e'))
                    mid_principal = float(match.group(3).replace('E', 'e'))
                    min_principal = float(match.group(4).replace('E', 'e'))
                    
                    # Calculate the sum of the principal stresses
                    sum_stresses = max_principal + mid_principal + min_principal
                    
                    # Store the values
                    element_labels.append(element_label)
                    sum_principal_stresses.append(sum_stresses)

    # Convert to numpy arrays for easier manipulation
    element_labels = np.array(element_labels)
    sum_principal_stresses = np.array(sum_principal_stresses)

    # Define the grid size (same for both regions)
    nrows, ncols = 28, 62

    # Generate the expected element labels grid based on the starting and ending element numbers
    expected_element_labels = np.arange(start_element, end_element + 1).reshape(nrows, ncols, order='F')

    # Create an array to store the principal stresses with NaN as default values
    reshaped_stress_data = np.full(expected_element_labels.shape, np.nan)

    # Place actual stress values into the correct positions, matching element numbers
    for i, elem in enumerate(element_labels):
        index = np.where(expected_element_labels == elem)
        reshaped_stress_data[index] = sum_principal_stresses[i]

    return reshaped_stress_data

# Function to calculate profiles including edge regions
def calculate_profiles_with_edges(reshaped_stress_data):
    # Central Horizontal Profile (rows 11 to 14)
    rows_to_average = reshaped_stress_data[10:14, :]
    horizontal_profile = np.nanmean(rows_to_average, axis=0)
    
    # Top Horizontal Profile (rows 1 to 3)
    top_rows = reshaped_stress_data[:3, :]
    horizontal_profile_top = np.nanmean(top_rows, axis=0)
    
    # Bottom Horizontal Profile (rows 26 to 28)
    bottom_rows = reshaped_stress_data[-3:, :]
    horizontal_profile_bottom = np.nanmean(bottom_rows, axis=0)

    # Central Vertical Profile (columns 29 to 32)
    columns_to_average = reshaped_stress_data[:, 28:32]
    vertical_profile = np.nanmean(columns_to_average, axis=1)
    
    # Left Vertical Profile (columns 1 to 3)
    left_columns = reshaped_stress_data[:, :3]
    vertical_profile_left = np.nanmean(left_columns, axis=1)
    
    # Right Vertical Profile (columns 60 to 62)
    right_columns = reshaped_stress_data[:, -3:]
    vertical_profile_right = np.nanmean(right_columns, axis=1)
    
    return (horizontal_profile, horizontal_profile_top, horizontal_profile_bottom,
            vertical_profile, vertical_profile_left, vertical_profile_right)

# Lists to store profiles for each layer and nodefect
horizontal_profiles = []
vertical_profiles = []

# Loop through each layer and extract profiles
for i in range(11, 20):
    file_path = f"ELCENT_defect_4x4_Layer{i}_Block45.rpt"
    start_element, end_element = 34721, 36456  # Element range for the region
    reshaped_stress_data = extract_data(file_path, start_element, end_element, f"INTERSTAGE_1/8-1.Region_21")
    
    # Calculate profiles
    horizontal_profile, vertical_profile = calculate_profiles(reshaped_stress_data)
    
    # Store profiles
    horizontal_profiles.append(horizontal_profile)
    vertical_profiles.append(vertical_profile)

# Add the Nodefect file for comparison
file_path_nodefect = "ELCENT_nodefect_origina.rpt"
start_element, end_element = 34721, 36456  # Same element range for nodefect
reshaped_stress_data_nodefect = extract_data(file_path_nodefect, start_element, end_element, "INTERSTAGE_1/8-1.Region_21")

# Calculate profiles for nodefect
horizontal_profile_nodefect, vertical_profile_nodefect = calculate_profiles(reshaped_stress_data_nodefect)

# Define width range (26 to 36)
width_range_start = 26
width_range_end = 36

# Extract the portion of the horizontal profiles corresponding to the width 26~36
horizontal_profiles_26_36 = [profile[width_range_start:width_range_end] for profile in horizontal_profiles]
horizontal_profile_nodefect_26_36 = horizontal_profile_nodefect[width_range_start:width_range_end]

# Plot horizontal profiles for the width range 26 to 36
plt.figure(figsize=(10, 6))
for i, profile in enumerate(horizontal_profiles_26_36, start=11):
    plt.plot(range(width_range_start, width_range_end), profile, label=f'Layer {i}')
plt.plot(range(width_range_start, width_range_end), horizontal_profile_nodefect_26_36, label='Nodefect', linestyle='--', color='black')
plt.title('Horizontal Profiles (Width 26 to 36)')
plt.xlabel('Width (Columns)')
plt.ylabel('Mean DSPSS')
plt.legend()
plt.show()
```

```python
# Horizontal Profiles: Calculate the difference between defect and nodefect profiles
horizontal_diff_profiles = [profile - horizontal_profile_nodefect for profile in horizontal_profiles]

# Normalize the differences to be in the range [0, 1], where smaller differences are closer to 1
normalized_diff_profiles = []
for diff_profile in horizontal_diff_profiles:
    max_diff = np.max(np.abs(diff_profile))  # Find the maximum absolute difference
    normalized_diff = 1 - np.abs(diff_profile) / max_diff  # Normalize the difference (inverted)
    normalized_diff_profiles.append(normalized_diff)

# Plot normalized difference profiles
plt.figure(figsize=(10, 6))

# Plot each normalized difference profile
for i, normalized_diff_profile in enumerate(normalized_diff_profiles, start=11):
    plt.plot(normalized_diff_profile, label=f'Layer {i} - Normalized Diff')

plt.title('Normalized Horizontal Differences (0 to 1 Scale)')
plt.xlabel('Width (Columns)')
plt.ylabel('Normalized Difference (Closer to 1 = Smaller Difference)')
plt.legend()
plt.show()
```

```python
# Horizontal Profiles: Calculate the difference between defect and nodefect profiles
horizontal_diff_profiles = [profile - horizontal_profile_nodefect for profile in horizontal_profiles]

# Define scaling factor based on 1 - (19 - i) * 0.05 for layers
normalized_diff_profiles = []
for i, diff_profile in enumerate(horizontal_diff_profiles, start=11):
    max_diff = np.max(np.abs(diff_profile))  # Find the maximum absolute difference
    normalized_diff = np.abs(diff_profile) / max_diff  # Normalize the difference
    
    # Apply scaling only for values greater than 0.2
    scaling_factor = 1 - (19 - i) * 0.1  # Scaling factor based on the layer
    adjusted_diff = np.where(normalized_diff > 0.2, normalized_diff * scaling_factor, normalized_diff)
    
    normalized_diff_profiles.append(adjusted_diff)

# Plot normalized difference profiles
plt.figure(figsize=(10, 6))

# Plot each normalized difference profile
for i, normalized_diff_profile in enumerate(normalized_diff_profiles, start=11):
    plt.plot(normalized_diff_profile, label=f'Layer {i} - Adjusted Diff')

plt.title('Adjusted Normalized Horizontal Differences with Layer-specific Scaling')
plt.xlabel('Width (Columns)')
plt.ylabel('Normalized Difference (Layer-specific Scaling)')
plt.legend()
plt.show()
```
