```python
import numpy as np

# 座標データのロード
x_coords = np.load("/home/nishioka/GNN/BasicdataforGNN/x_2layer.npy")[:3654]
y_coords = np.load("/home/nishioka/GNN/BasicdataforGNN/y_2layer.npy")[:3654]
z_coords = np.load("/home/nishioka/GNN/BasicdataforGNN/z_2layer.npy")[:3654]

# 正規化用の関数
def normalize(coords):
    min_val = np.min(coords)
    max_val = np.max(coords)
    return (coords - min_val) / (max_val - min_val)

# 標準化用の関数
def standardize(coords):
    mean_val = np.mean(coords)
    std_val = np.std(coords)
    return (coords - mean_val) / std_val

# x, y, z 座標の正規化
x_coords_normalized = normalize(x_coords)
y_coords_normalized = normalize(y_coords)
z_coords_normalized = normalize(z_coords)

# x, y, z 座標の標準化
x_coords_standardized = standardize(x_coords)
y_coords_standardized = standardize(y_coords)
z_coords_standardized = standardize(z_coords)

# ファイルに保存
np.save('/home/nishioka/GNN/BasicdataforGNN/x_2layer_normalized.npy', x_coords_normalized)
np.save('/home/nishioka/GNN/BasicdataforGNN/y_2layer_normalized.npy', y_coords_normalized)
np.save('/home/nishioka/GNN/BasicdataforGNN/z_2layer_normalized.npy', z_coords_normalized)

np.save('/home/nishioka/GNN/BasicdataforGNN/x_2layer_standardized.npy', x_coords_standardized)
np.save('/home/nishioka/GNN/BasicdataforGNN/y_2layer_standardized.npy', y_coords_standardized)
np.save('/home/nishioka/GNN/BasicdataforGNN/z_2layer_standardized.npy', z_coords_standardized)

"/mnt/data/ファイルに保存しました"
```

```python
import matplotlib.pyplot as plt

# 正規化および標準化データを使って分布をプロットする関数
def plot_distribution(data, title, subplot_position):
    plt.subplot(subplot_position)
    plt.hist(data, bins=50, alpha=0.75, color='blue', edgecolor='black')
    plt.title(title)
    plt.xlabel('Value')
    plt.ylabel('Frequency')

# プロットの作成
plt.figure(figsize=(15, 10))

# x, y, z 座標の正規化後の分布
plot_distribution(x_coords_normalized, "X Coordinates (Normalized)", 231)
plot_distribution(y_coords_normalized, "Y Coordinates (Normalized)", 232)
plot_distribution(z_coords_normalized, "Z Coordinates (Normalized)", 233)

# x, y, z 座標の標準化後の分布
plot_distribution(x_coords_standardized, "X Coordinates (Standardized)", 234)
plot_distribution(y_coords_standardized, "Y Coordinates (Standardized)", 235)
plot_distribution(z_coords_standardized, "Z Coordinates (Standardized)", 236)

plt.tight_layout()
plt.show()
```
